```python
# ============================================================
# CatBoost
# HTS vs LTS
# Nested donor-aware cross-validation
#
# Feature ranking:
# donor-balanced mean(|SHAP|)
#
# HTS = 1
# LTS = 0
# ============================================================


# ============================================================
# 0. Import
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from catboost import (
    CatBoostClassifier,
    Pool
)

from sklearn.model_selection import GroupKFold

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    roc_curve,
    precision_recall_curve
)


# ============================================================
# 1. Parameters
# ============================================================

DATA_FILE = "AD_microglia_HTS_LTS_TWAS_genes_ML.csv"

FIG_DIR = "figures"
RESULT_DIR = "results"

os.makedirs(
    FIG_DIR,
    exist_ok=True
)

os.makedirs(
    RESULT_DIR,
    exist_ok=True
)


RANDOM_STATE = 42

TOP_N = 20


# ============================================================
# 2. CatBoost parameter grid
#
# 2 × 2 × 2 = 8 combinations
#
# iterations:
# number of boosting iterations
#
# learning_rate:
# contribution of each tree
#
# depth:
# tree depth
#
# 不把参数网格设得过大，
# 因为还要做9 outer × 4 inner
# ============================================================

CAT_PARAM_GRID = []


for iterations in [
    100,
    200
]:

    for learning_rate in [
        0.03,
        0.1
    ]:

        for depth in [
            4,
            6
        ]:

            CAT_PARAM_GRID.append({

                "iterations":
                    iterations,

                "learning_rate":
                    learning_rate,

                "depth":
                    depth

            })


print(
    "Number of CatBoost parameter combinations:",
    len(CAT_PARAM_GRID)
)


# ============================================================
# 3. Read data
# ============================================================

df = pd.read_csv(
    DATA_FILE
)


print(
    "Original data shape:",
    df.shape
)

print(
    df.head()
)


# ============================================================
# 4. Handle cell barcode
# ============================================================

first_col = df.columns[0]


if first_col.startswith(
    "Unnamed"
):

    df = df.rename(

        columns={
            first_col:
                "cell"
        }

    )

    df = df.set_index(
        "cell"
    )


elif first_col.lower() in [

    "cell",
    "barcode",
    "cell_barcode"

]:

    df = df.set_index(
        first_col
    )


else:

    df.index = (
        df.index
        .astype(str)
    )

    df.index.name = "cell"


print(
    "\nData after setting cell index:"
)

print(
    df.head()
)


# ============================================================
# 5. Required columns
# ============================================================

required_columns = [

    "individual",
    "label"

]


for col in required_columns:

    if col not in df.columns:

        raise ValueError(
            f"Missing required column: {col}"
        )


# ============================================================
# 6. Metadata
# ============================================================

df[
    "individual"
] = (

    df[
        "individual"
    ]

    .astype(str)

)


df[
    "label"
] = (

    df[
        "label"
    ]

    .astype(int)

)


print(
    "\nLabel distribution:"
)

print(

    df[
        "label"
    ]

    .value_counts()

)


print(
    "\nDonor × label:"
)

print(

    pd.crosstab(

        df[
            "individual"
        ],

        df[
            "label"
        ]

    )

)


# ============================================================
# 7. Define genes
#
# individual = donor grouping only
# label      = target
#
# remaining columns = TWAS genes
# ============================================================

gene_columns = [

    col

    for col in df.columns

    if col not in [

        "individual",
        "label"

    ]

]


X = (

    df[
        gene_columns
    ]

    .astype(float)

)


y = (

    df[
        "label"
    ]

    .values

)


groups = (

    df[
        "individual"
    ]

    .values

)


cells = (
    df.index.values
)


print(
    "\nCells:",
    X.shape[0]
)

print(
    "Genes:",
    X.shape[1]
)

print(
    "Donors:",
    len(
        np.unique(
            groups
        )
    )
)


# ============================================================
# 8. Basic checks
# ============================================================

n_missing = (

    X
    .isna()
    .sum()
    .sum()

)


print(
    "Missing values:",
    n_missing
)


if n_missing > 0:

    raise ValueError(
        "Expression matrix contains missing values."
    )


# ------------------------------------------------------------

n_inf = (

    np.isinf(
        X.to_numpy()
    )

    .sum()

)


print(
    "Infinite values:",
    n_inf
)


if n_inf > 0:

    raise ValueError(
        "Expression matrix contains infinite values."
    )


# ------------------------------------------------------------

n_duplicate_genes = (

    pd.Index(
        gene_columns
    )

    .duplicated()
    .sum()

)


print(
    "Duplicated gene columns:",
    n_duplicate_genes
)


# ------------------------------------------------------------

zero_variance_genes = (

    X.columns[
        X.var(
            axis=0
        ) == 0
    ]

    .tolist()

)


print(
    "Zero variance genes:",
    len(
        zero_variance_genes
    )
)


if len(
    zero_variance_genes
) > 0:

    print(
        zero_variance_genes
    )


# ============================================================
# 9. Donor check
# ============================================================

donors = np.sort(

    np.unique(
        groups
    )

)


print(
    "\nDonors:"
)

print(
    donors
)


# 每个donor必须同时包含HTS和LTS
for donor in donors:

    donor_y = y[
        groups == donor
    ]

    if len(
        np.unique(
            donor_y
        )
    ) < 2:

        raise ValueError(

            f"Donor {donor} "
            "does not contain both HTS and LTS."

        )


# ============================================================
# 10. donor × class balanced sample weights
#
# 目的：
#
# donor A × HTS
# donor A × LTS
# donor B × HTS
# donor B × LTS
#
# 每个组合对training的总贡献近似相等
# ============================================================

def make_donor_class_weights(
    groups,
    y
):

    temp = pd.DataFrame({

        "individual":
            groups,

        "label":
            y

    })


    counts = (

        temp

        .groupby(
            [
                "individual",
                "label"
            ]
        )

        .size()

    )


    weights = []


    for donor, label in zip(
        groups,
        y
    ):

        n = counts.loc[
            (
                donor,
                label
            )
        ]


        weights.append(
            1.0 / n
        )


    weights = np.array(
        weights
    )


    # Normalize to mean weight = 1
    weights = (

        weights /
        weights.mean()

    )


    return weights


# ============================================================
# 11. Build CatBoost model
#
# 不需要StandardScaler
#
# allow_writing_files=False：
# 防止CatBoost自动生成catboost_info目录
# ============================================================

def build_cat_model(
    params
):

    model = CatBoostClassifier(

        loss_function=
            "Logloss",

        eval_metric=
            "Logloss",

        iterations=
            params[
                "iterations"
            ],

        learning_rate=
            params[
                "learning_rate"
            ],

        depth=
            params[
                "depth"
            ],

        l2_leaf_reg=
            3.0,

        random_seed=
            RANDOM_STATE,

        verbose=
            False,

        allow_writing_files=
            False,

        thread_count=
            -1

    )


    return model


# ============================================================
# 12. Donor-level expression direction
#
# 注意：
#
# SHAP = 模型贡献
#
# expression direction =
# 实际基因表达 HTS - LTS
#
# 两者不是同一个概念
# ============================================================

def donor_expression_direction(
    X_input,
    y_input,
    groups_input
):

    donor_deltas = []


    for donor in np.unique(
        groups_input
    ):

        donor_mask = (

            groups_input
            ==
            donor

        )


        donor_y = (

            y_input[
                donor_mask
            ]

        )


        if len(
            np.unique(
                donor_y
            )
        ) < 2:

            continue


        X_donor = (

            X_input.loc[
                donor_mask
            ]

        )


        # ----------------------------------------------------
        # mean expression in HTS
        # ----------------------------------------------------

        mean_hts = (

            X_donor.loc[
                donor_y == 1
            ]

            .mean(
                axis=0
            )

        )


        # ----------------------------------------------------
        # mean expression in LTS
        # ----------------------------------------------------

        mean_lts = (

            X_donor.loc[
                donor_y == 0
            ]

            .mean(
                axis=0
            )

        )


        delta = (

            mean_hts
            -
            mean_lts

        )


        donor_deltas.append(
            delta
        )


    # ========================================================
    # donor × gene
    # ========================================================

    delta_df = pd.DataFrame(
        donor_deltas
    )


    # ========================================================
    # mean donor-level HTS-LTS
    # ========================================================

    mean_delta = (

        delta_df

        .mean(
            axis=0
        )

    )


    # ========================================================
    # direction consistency
    # ========================================================

    positive_fraction = (

        (
            delta_df > 0
        )

        .mean(
            axis=0
        )

    )


    negative_fraction = (

        (
            delta_df < 0
        )

        .mean(
            axis=0
        )

    )


    direction_consistency = (

        pd.concat(

            [
                positive_fraction,
                negative_fraction
            ],

            axis=1

        )

        .max(
            axis=1
        )

    )


    result = pd.DataFrame({

        "gene":
            X_input.columns,

        "mean_HTS_minus_LTS":

            mean_delta[
                X_input.columns
            ].values,

        "expression_direction_consistency":

            direction_consistency[
                X_input.columns
            ].values

    })


    result[
        "expression_direction"
    ] = np.where(

        result[
            "mean_HTS_minus_LTS"
        ] > 0,

        "HTS_high",

        "LTS_high"

    )


    return result


# ============================================================
# 13. CatBoost donor-balanced SHAP
#
# CatBoost ShapValues返回：
#
# n_cells × (n_genes + 1)
#
# 最后一列 = expected value
#
# 所以前n_genes列才是真正的gene SHAP values
#
#
# 为避免cell多的donor主导：
#
# donor1 → mean(|SHAP|)
# donor2 → mean(|SHAP|)
# ...
# donor8 → mean(|SHAP|)
#
# 最后donor等权平均
# ============================================================

def donor_balanced_shap_importance(

    model,

    X_input,

    groups_input

):

    # ========================================================
    # Build CatBoost Pool
    # ========================================================

    pool = Pool(

        data=
            X_input,

        feature_names=
            list(
                X_input.columns
            )

    )


    # ========================================================
    # CatBoost native SHAP
    # ========================================================

    shap_matrix = (

        model

        .get_feature_importance(

            data=
                pool,

            type=
                "ShapValues",

            thread_count=
                -1,

            verbose=
                False

        )

    )


    shap_matrix = np.asarray(
        shap_matrix
    )


    # ========================================================
    # Shape check
    #
    # binary classification:
    #
    # n_cells × (n_genes + 1)
    # ========================================================

    if shap_matrix.ndim != 2:

        raise ValueError(

            "Unexpected CatBoost SHAP dimension: "
            f"{shap_matrix.shape}"

        )


    expected_columns = (
        X_input.shape[1] + 1
    )


    if shap_matrix.shape[1] != expected_columns:

        raise ValueError(

            "Unexpected CatBoost SHAP shape: "
            f"{shap_matrix.shape}; "
            f"expected {expected_columns} columns."

        )


    # ========================================================
    # Last column = expected value
    # ========================================================

    shap_values = (

        shap_matrix[
            :,
            :-1
        ]

    )


    # ========================================================
    # donor-wise SHAP
    # ========================================================

    donor_abs_shap = []

    donor_signed_shap = []


    donor_names = np.unique(
        groups_input
    )


    for donor in donor_names:


        donor_mask = (

            groups_input
            ==
            donor

        )


        donor_shap = (

            shap_values[
                donor_mask,
                :
            ]

        )


        # ----------------------------------------------------
        # importance
        # ----------------------------------------------------

        donor_abs = (

            np.mean(

                np.abs(
                    donor_shap
                ),

                axis=0

            )

        )


        # ----------------------------------------------------
        # signed contribution
        #
        # 仅用于模型贡献参考
        # 不代表真实表达上/下调
        # ----------------------------------------------------

        donor_signed = (

            np.mean(

                donor_shap,

                axis=0

            )

        )


        donor_abs_shap.append(
            donor_abs
        )


        donor_signed_shap.append(
            donor_signed
        )


    donor_abs_shap = np.vstack(
        donor_abs_shap
    )


    donor_signed_shap = np.vstack(
        donor_signed_shap
    )


    # ========================================================
    # Equal-weight donors
    # ========================================================

    mean_abs_shap = (

        donor_abs_shap

        .mean(
            axis=0
        )

    )


    median_abs_shap = (

        np.median(

            donor_abs_shap,

            axis=0

        )

    )


    sd_abs_shap = (

        donor_abs_shap

        .std(
            axis=0,
            ddof=1
        )

    )


    mean_signed_shap = (

        donor_signed_shap

        .mean(
            axis=0
        )

    )


    result = pd.DataFrame({

        "gene":
            X_input.columns,

        "mean_abs_SHAP":
            mean_abs_shap,

        "median_abs_SHAP":
            median_abs_shap,

        "sd_abs_SHAP":
            sd_abs_shap,

        "mean_signed_SHAP":
            mean_signed_shap

    })


    return result


# ============================================================
# 14. Result containers
# ============================================================

outer_metrics = []

all_predictions = []

all_shap_importance = []

all_top20 = []

all_tuning_results = []


# ============================================================
# 15. Outer Leave-One-Donor-Out
# ============================================================

print(
    "\nNumber of outer folds:",
    len(donors)
)


for fold_number, test_donor in enumerate(

    donors,

    start=1

):


    print("\n")

    print(
        "=" * 90
    )


    print(

        f"Outer fold "
        f"{fold_number}/"
        f"{len(donors)}"

    )


    print(
        f"Test donor: {test_donor}"
    )


    print(
        "=" * 90
    )


    # ========================================================
    # Outer development / test split
    # ========================================================

    test_mask = (

        groups
        ==
        test_donor

    )


    dev_mask = (
        ~test_mask
    )


    X_dev = (

        X.loc[
            dev_mask
        ]

        .reset_index(
            drop=True
        )

    )


    y_dev = (

        y[
            dev_mask
        ]

    )


    groups_dev = (

        groups[
            dev_mask
        ]

    )


    X_test = (

        X.loc[
            test_mask
        ]

        .reset_index(
            drop=True
        )

    )


    y_test = (

        y[
            test_mask
        ]

    )


    test_cells = (

        cells[
            test_mask
        ]

    )


    print(

        "Development donors:",

        len(
            np.unique(
                groups_dev
            )
        )

    )


    print(
        "Development cells:",
        len(
            y_dev
        )
    )


    print(
        "Test cells:",
        len(
            y_test
        )
    )


    # ========================================================
    # 16. Inner 4-fold donor CV
    # ========================================================

    inner_cv = GroupKFold(
        n_splits=4
    )


    parameter_scores = []


    # ========================================================
    # Parameter tuning
    # ========================================================

    for params in CAT_PARAM_GRID:


        inner_AP = []

        inner_ROC = []


        # ----------------------------------------------------
        # Inner donor CV
        # ----------------------------------------------------

        for (

            inner_train_idx,
            inner_val_idx

        ) in inner_cv.split(

            X_dev,

            y_dev,

            groups_dev

        ):


            # =================================================
            # Inner training
            # =================================================

            X_inner_train = (

                X_dev.iloc[
                    inner_train_idx
                ]

            )


            y_inner_train = (

                y_dev[
                    inner_train_idx
                ]

            )


            groups_inner_train = (

                groups_dev[
                    inner_train_idx
                ]

            )


            # =================================================
            # Inner validation
            # =================================================

            X_inner_val = (

                X_dev.iloc[
                    inner_val_idx
                ]

            )


            y_inner_val = (

                y_dev[
                    inner_val_idx
                ]

            )


            groups_inner_val = (

                groups_dev[
                    inner_val_idx
                ]

            )


            # =================================================
            # donor × class weights
            # =================================================

            train_weights = (

                make_donor_class_weights(

                    groups_inner_train,

                    y_inner_train

                )

            )


            # =================================================
            # Train CatBoost
            # =================================================

            model = build_cat_model(
                params
            )


            model.fit(

                X_inner_train,

                y_inner_train,

                sample_weight=
                    train_weights,

                verbose=
                    False

            )


            # =================================================
            # Validation probability
            # =================================================

            val_prob = (

                model

                .predict_proba(
                    X_inner_val
                )[:, 1]

            )


            # =================================================
            # donor-macro evaluation
            #
            # validation donor分别计算AP/AUC
            #
            # 再进行donor等权平均
            # =================================================

            val_donors = np.unique(
                groups_inner_val
            )


            for val_donor in val_donors:


                donor_mask = (

                    groups_inner_val
                    ==
                    val_donor

                )


                donor_y = (

                    y_inner_val[
                        donor_mask
                    ]

                )


                donor_prob = (

                    val_prob[
                        donor_mask
                    ]

                )


                if len(
                    np.unique(
                        donor_y
                    )
                ) < 2:

                    continue


                donor_AP = (

                    average_precision_score(

                        donor_y,

                        donor_prob

                    )

                )


                donor_ROC = (

                    roc_auc_score(

                        donor_y,

                        donor_prob

                    )

                )


                inner_AP.append(
                    donor_AP
                )


                inner_ROC.append(
                    donor_ROC
                )


        # ====================================================
        # Mean donor-level validation performance
        # ====================================================

        mean_AP = (

            np.mean(
                inner_AP
            )

        )


        mean_ROC = (

            np.mean(
                inner_ROC
            )

        )


        parameter_scores.append({

            "iterations":

                params[
                    "iterations"
                ],

            "learning_rate":

                params[
                    "learning_rate"
                ],

            "depth":

                params[
                    "depth"
                ],

            "AP":

                mean_AP,

            "ROC_AUC":

                mean_ROC

        })


        all_tuning_results.append({

            "outer_test_donor":

                test_donor,

            "iterations":

                params[
                    "iterations"
                ],

            "learning_rate":

                params[
                    "learning_rate"
                ],

            "depth":

                params[
                    "depth"
                ],

            "inner_macro_AP":

                mean_AP,

            "inner_macro_ROC_AUC":

                mean_ROC

        })


    # ========================================================
    # 17. Select best CatBoost parameters
    #
    # Primary:
    # AP
    #
    # Secondary:
    # ROC-AUC
    #
    # tie:
    # shallower model preferred
    # ========================================================

    parameter_scores_df = pd.DataFrame(
        parameter_scores
    )


    best_row = (

        parameter_scores_df

        .sort_values(

            [
                "AP",
                "ROC_AUC",
                "depth",
                "iterations"
            ],

            ascending=[
                False,
                False,
                True,
                True
            ]

        )

        .iloc[0]

    )


    best_params = {

        "iterations":

            int(
                best_row[
                    "iterations"
                ]
            ),

        "learning_rate":

            float(
                best_row[
                    "learning_rate"
                ]
            ),

        "depth":

            int(
                best_row[
                    "depth"
                ]
            )

    }


    print(
        "\nBest CatBoost parameters:"
    )


    print(
        best_params
    )


    print(

        "Inner macro AP:",

        round(
            best_row[
                "AP"
            ],
            4
        )

    )


    print(

        "Inner macro ROC-AUC:",

        round(
            best_row[
                "ROC_AUC"
            ],
            4
        )

    )


    # ========================================================
    # 18. Retrain using all 8 development donors
    # ========================================================

    dev_weights = (

        make_donor_class_weights(

            groups_dev,

            y_dev

        )

    )


    final_model = build_cat_model(
        best_params
    )


    final_model.fit(

        X_dev,

        y_dev,

        sample_weight=
            dev_weights,

        verbose=
            False

    )


    # ========================================================
    # 19. Predict completely held-out donor
    #
    # Outer test donor第一次进入模型评价
    # ========================================================

    test_prob = (

        final_model

        .predict_proba(
            X_test
        )[:, 1]

    )


    test_pred = (

        test_prob
        >= 0.5

    ).astype(
        int
    )


    # ========================================================
    # 20. Outer test performance
    # ========================================================

    test_ROC = (

        roc_auc_score(

            y_test,

            test_prob

        )

    )


    test_AP = (

        average_precision_score(

            y_test,

            test_prob

        )

    )


    print(

        "Outer ROC-AUC:",

        round(
            test_ROC,
            4
        )

    )


    print(

        "Outer AP:",

        round(
            test_AP,
            4
        )

    )


    outer_metrics.append({

        "fold":

            fold_number,

        "test_donor":

            test_donor,

        "best_iterations":

            best_params[
                "iterations"
            ],

        "best_learning_rate":

            best_params[
                "learning_rate"
            ],

        "best_depth":

            best_params[
                "depth"
            ],

        "n_test_cells":

            len(
                y_test
            ),

        "n_HTS":

            int(
                np.sum(
                    y_test == 1
                )
            ),

        "n_LTS":

            int(
                np.sum(
                    y_test == 0
                )
            ),

        "ROC_AUC":

            test_ROC,

        "AP":

            test_AP

    })


    # ========================================================
    # 21. Save outer predictions
    # ========================================================

    prediction_df = pd.DataFrame({

        "cell":

            test_cells,

        "individual":

            test_donor,

        "true_label":

            y_test,

        "predicted_probability":

            test_prob,

        "predicted_label":

            test_pred

    })


    all_predictions.append(
        prediction_df
    )


    # ========================================================
    # 22. CatBoost native SHAP
    #
    # IMPORTANT:
    #
    # 只使用8个development donors
    #
    # 不使用outer test donor筛基因
    # ========================================================

    print(
        "Calculating donor-balanced CatBoost SHAP..."
    )


    shap_df = (

        donor_balanced_shap_importance(

            final_model,

            X_dev,

            groups_dev

        )

    )


    shap_df[
        "outer_test_donor"
    ] = test_donor


    # ========================================================
    # 23. Actual expression direction
    # ========================================================

    direction_df = (

        donor_expression_direction(

            X_dev,

            y_dev,

            groups_dev

        )

    )


    shap_df = (

        shap_df

        .merge(

            direction_df,

            on=
                "gene",

            how=
                "left"

        )

    )


    all_shap_importance.append(
        shap_df
    )


    # ========================================================
    # 24. Top20 genes
    #
    # Ranking:
    #
    # donor-balanced mean(|SHAP|)
    # ========================================================

    top20 = (

        shap_df

        .sort_values(

            "mean_abs_SHAP",

            ascending=False

        )

        .head(
            TOP_N
        )

        .copy()

    )


    top20[
        "rank"
    ] = np.arange(

        1,

        len(
            top20
        ) + 1

    )


    all_top20.append(
        top20
    )


    print(
        "\nTop 10 CatBoost genes:"
    )


    print(

        top20[
            [
                "gene",
                "mean_abs_SHAP",
                "median_abs_SHAP",
                "mean_HTS_minus_LTS",
                "expression_direction"
            ]
        ]

        .head(10)

        .to_string(
            index=False
        )

    )


# ============================================================
# 25. Combine results
# ============================================================

outer_metrics_df = pd.DataFrame(
    outer_metrics
)


predictions_df = pd.concat(

    all_predictions,

    ignore_index=True

)


shap_importance_df = pd.concat(

    all_shap_importance,

    ignore_index=True

)


top20_df = pd.concat(

    all_top20,

    ignore_index=True

)


tuning_df = pd.DataFrame(
    all_tuning_results
)


# ============================================================
# 26. Save raw results
# ============================================================

outer_metrics_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "CAT_outer_LODO_performance.csv"

    ),

    index=False

)


predictions_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "CAT_outer_LODO_predictions.csv"

    ),

    index=False

)


shap_importance_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "CAT_all_fold_SHAP_importance.csv"

    ),

    index=False

)


top20_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "CAT_top20_each_fold.csv"

    ),

    index=False

)


tuning_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "CAT_inner_parameter_tuning.csv"

    ),

    index=False

)


# ============================================================
# 27. Top20 frequency
#
# 一个gene在9个outer folds中
# 多少次进入Top20
# ============================================================

top20_frequency = (

    top20_df

    .groupby(
        "gene"
    )

    .agg(

        top20_count=(

            "outer_test_donor",

            "nunique"

        )

    )

    .reset_index()

)


# ============================================================
# 28. SHAP summary across 9 outer folds
# ============================================================

shap_summary = (

    shap_importance_df

    .groupby(
        "gene"
    )

    .agg(

        mean_abs_SHAP=(

            "mean_abs_SHAP",

            "mean"

        ),

        median_abs_SHAP=(

            "mean_abs_SHAP",

            "median"

        ),

        sd_abs_SHAP=(

            "mean_abs_SHAP",

            "std"

        ),

        mean_signed_SHAP=(

            "mean_signed_SHAP",

            "mean"

        ),

        mean_HTS_minus_LTS=(

            "mean_HTS_minus_LTS",

            "mean"

        ),

        mean_expression_direction_consistency=(

            "expression_direction_consistency",

            "mean"

        )

    )

    .reset_index()

)


# ============================================================
# 29. Merge stability
# ============================================================

feature_summary = (

    shap_summary

    .merge(

        top20_frequency,

        on=
            "gene",

        how=
            "left"

    )

)


feature_summary[
    "top20_count"
] = (

    feature_summary[
        "top20_count"
    ]

    .fillna(0)

    .astype(int)

)


# ============================================================
# 30. Stable CatBoost gene
#
# Top20 in >= 6 / 9 outer folds
# ============================================================

feature_summary[
    "stable_top20"
] = (

    feature_summary[
        "top20_count"
    ]

    >= 6

)


# ============================================================
# 31. Actual expression direction
# ============================================================

feature_summary[
    "expression_direction"
] = np.where(

    feature_summary[
        "mean_HTS_minus_LTS"
    ] > 0,

    "HTS_high",

    "LTS_high"

)


# ============================================================
# 32. Final ranking
#
# Priority 1:
# Top20 stability
#
# Priority 2:
# mean(|SHAP|)
# ============================================================

feature_summary = (

    feature_summary

    .sort_values(

        [
            "top20_count",
            "mean_abs_SHAP"
        ],

        ascending=[
            False,
            False
        ]

    )

    .reset_index(
        drop=True
    )

)


feature_summary[
    "final_rank"
] = np.arange(

    1,

    len(
        feature_summary
    ) + 1

)


# ============================================================
# 33. Save final feature summary
# ============================================================

feature_summary.to_csv(

    os.path.join(

        RESULT_DIR,

        "CAT_feature_stability_summary.csv"

    ),

    index=False

)


# ============================================================
# 34. Print CatBoost feature ranking
# ============================================================

print("\n")

print(
    "=" * 90
)

print(
    "CatBoost feature stability ranking"
)

print(
    "=" * 90
)


print(

    feature_summary[
        [
            "final_rank",
            "gene",
            "top20_count",
            "mean_abs_SHAP",
            "median_abs_SHAP",
            "mean_HTS_minus_LTS",
            "expression_direction",
            "mean_expression_direction_consistency",
            "stable_top20"
        ]
    ]

    .head(30)

    .to_string(
        index=False
    )

)


# ============================================================
# 35. Outer performance summary
# ============================================================

roc_median = (

    outer_metrics_df[
        "ROC_AUC"
    ]

    .median()

)


roc_q1 = (

    outer_metrics_df[
        "ROC_AUC"
    ]

    .quantile(
        0.25
    )

)


roc_q3 = (

    outer_metrics_df[
        "ROC_AUC"
    ]

    .quantile(
        0.75
    )

)


ap_median = (

    outer_metrics_df[
        "AP"
    ]

    .median()

)


ap_q1 = (

    outer_metrics_df[
        "AP"
    ]

    .quantile(
        0.25
    )

)


ap_q3 = (

    outer_metrics_df[
        "AP"
    ]

    .quantile(
        0.75
    )

)


print("\n")

print(
    "=" * 90
)

print(
    "Outer Leave-One-Donor-Out Performance"
)

print(
    "=" * 90
)


print(

    outer_metrics_df

    .to_string(
        index=False
    )

)


print(
    "\nROC-AUC median [IQR]:"
)


print(

    f"{roc_median:.3f} "
    f"[{roc_q1:.3f}, {roc_q3:.3f}]"

)


print(
    "\nAP median [IQR]:"
)


print(

    f"{ap_median:.3f} "
    f"[{ap_q1:.3f}, {ap_q3:.3f}]"

)


# ============================================================
# 36. Save performance summary
# ============================================================

performance_summary = pd.DataFrame({

    "metric": [
        "ROC_AUC",
        "AP"
    ],

    "median": [
        roc_median,
        ap_median
    ],

    "Q1": [
        roc_q1,
        ap_q1
    ],

    "Q3": [
        roc_q3,
        ap_q3
    ]

})


performance_summary.to_csv(

    os.path.join(

        RESULT_DIR,

        "CAT_performance_summary.csv"

    ),

    index=False

)


# ============================================================
# 37. Pooled OOF ROC
#
# visualization only
#
# 正式performance优先：
# 9 donor median [IQR]
# ============================================================

y_oof = (

    predictions_df[
        "true_label"
    ]

    .values

)


prob_oof = (

    predictions_df[
        "predicted_probability"
    ]

    .values

)


fpr, tpr, _ = roc_curve(

    y_oof,

    prob_oof

)


pooled_roc = (

    roc_auc_score(

        y_oof,

        prob_oof

    )

)


plt.figure(
    figsize=(
        6,
        6
    )
)


plt.plot(

    fpr,

    tpr,

    linewidth=2,

    label=(
        f"CatBoost AUC = "
        f"{pooled_roc:.3f}"
    )

)


plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--"

)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "CatBoost\nLeave-One-Donor-Out ROC"
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "CAT_LODO_ROC.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 38. Precision-Recall
# ============================================================

precision, recall, _ = (

    precision_recall_curve(

        y_oof,

        prob_oof

    )

)


pooled_ap = (

    average_precision_score(

        y_oof,

        prob_oof

    )

)


plt.figure(
    figsize=(
        6,
        6
    )
)


plt.plot(

    recall,

    precision,

    linewidth=2,

    label=(
        f"CatBoost AP = "
        f"{pooled_ap:.3f}"
    )

)


plt.xlabel(
    "Recall"
)

plt.ylabel(
    "Precision"
)

plt.title(
    "CatBoost\n"
    "Leave-One-Donor-Out Precision-Recall"
)

plt.legend(
    loc="lower left"
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "CAT_LODO_PR.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 39. Feature stability plot
# ============================================================

plot_df = (

    feature_summary

    .head(
        20
    )

    .copy()

)


plot_df = (

    plot_df

    .sort_values(

        "top20_count",

        ascending=True

    )

)


plt.figure(
    figsize=(
        7,
        8
    )
)


plt.barh(

    plot_df[
        "gene"
    ],

    plot_df[
        "top20_count"
    ]

)


plt.xlabel(
    "Number of outer folds in Top 20"
)

plt.ylabel(
    "Gene"
)

plt.title(
    "CatBoost SHAP Feature Stability"
)

plt.xlim(
    0,
    len(donors)
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "CAT_Top20_feature_stability.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 40. Top20 SHAP magnitude
# ============================================================

shap_plot_df = (

    feature_summary

    .head(
        20
    )

    .sort_values(

        "mean_abs_SHAP",

        ascending=True

    )

)


plt.figure(
    figsize=(
        7,
        8
    )
)


plt.barh(

    shap_plot_df[
        "gene"
    ],

    shap_plot_df[
        "mean_abs_SHAP"
    ]

)


plt.xlabel(
    "Mean absolute SHAP value"
)

plt.ylabel(
    "Gene"
)

plt.title(
    "CatBoost Donor-Balanced SHAP Importance"
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "CAT_Top20_mean_abs_SHAP.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 41. Donor-level performance plot
# ============================================================

performance_plot = (

    outer_metrics_df

    .sort_values(
        "ROC_AUC"
    )

    .copy()

)


x = np.arange(
    len(
        performance_plot
    )
)


width = 0.35


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.bar(

    x - width / 2,

    performance_plot[
        "ROC_AUC"
    ],

    width,

    label=
        "ROC-AUC"

)


plt.bar(

    x + width / 2,

    performance_plot[
        "AP"
    ],

    width,

    label=
        "AP"

)


plt.xticks(

    x,

    performance_plot[
        "test_donor"
    ],

    rotation=45,

    ha="right"

)


plt.ylim(
    0,
    1
)

plt.ylabel(
    "Performance"
)

plt.xlabel(
    "Held-out donor"
)

plt.title(
    "CatBoost Performance by Held-out Donor"
)

plt.legend()

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "CAT_outer_donor_performance.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 42. Print stable genes
# ============================================================

stable_genes = (

    feature_summary[

        feature_summary[
            "stable_top20"
        ]

    ]

)


print("\n")

print(
    "=" * 90
)

print(
    "Stable CatBoost genes "
    "(Top20 in >= 6/9 folds)"
)

print(
    "=" * 90
)


if len(
    stable_genes
) == 0:

    print(

        "No genes reached "
        ">= 6/9 Top20 stability."

    )


else:

    print(

        stable_genes[
            [
                "gene",
                "top20_count",
                "mean_abs_SHAP",
                "median_abs_SHAP",
                "mean_HTS_minus_LTS",
                "expression_direction",
                "mean_expression_direction_consistency"
            ]
        ]

        .to_string(
            index=False
        )

    )


# ============================================================
# 43. Done
# ============================================================

print(
    "\nAnalysis completed."
)


print(

    f"\nFigures saved to: "
    f"{FIG_DIR}/"

)


print(

    f"Results saved to: "
    f"{RESULT_DIR}/"

)
```
