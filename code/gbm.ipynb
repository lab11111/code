```python
# ============================================================
# Gradient Boosting Machine (GBM)
# HTS vs LTS
# Nested donor-aware cross-validation
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import GroupKFold
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    roc_curve,
    precision_recall_curve
)


# ============================================================
# 1. Parameters
# ============================================================

DATA_FILE = "AD_microglia_HTS_LTS_TWAS_genes_ML.csv"

FIG_DIR = "figures"
RESULT_DIR = "results"

os.makedirs(FIG_DIR, exist_ok=True)
os.makedirs(RESULT_DIR, exist_ok=True)

RANDOM_STATE = 42

TOP_N = 20


# ============================================================
# 2. GBM parameter grid
#
# 12 combinations
#
# n_estimators:
# number of boosting trees
#
# learning_rate:
# contribution of each tree
#
# max_depth:
# complexity of individual trees
# ============================================================

GBM_PARAM_GRID = []

for n_estimators in [100, 200]:

    for learning_rate in [0.03, 0.1]:

        for max_depth in [1, 2, 3]:

            GBM_PARAM_GRID.append({

                "n_estimators":
                    n_estimators,

                "learning_rate":
                    learning_rate,

                "max_depth":
                    max_depth

            })


print(
    "Number of GBM parameter combinations:",
    len(GBM_PARAM_GRID)
)


# ============================================================
# 3. Read data
# ============================================================

df = pd.read_csv(DATA_FILE)

print(
    "Original data shape:",
    df.shape
)

print(
    df.head()
)


# ============================================================
# 4. Handle cell barcode
# ============================================================

first_col = df.columns[0]


if first_col.startswith("Unnamed"):

    df = df.rename(
        columns={
            first_col: "cell"
        }
    )

    df = df.set_index(
        "cell"
    )


elif first_col.lower() in [
    "cell",
    "barcode",
    "cell_barcode"
]:

    df = df.set_index(
        first_col
    )


else:

    df.index = (
        df.index
        .astype(str)
    )

    df.index.name = "cell"


print(
    "\nData after setting cell index:"
)

print(
    df.head()
)


# ============================================================
# 5. Check required columns
# ============================================================

required_columns = [
    "individual",
    "label"
]


for col in required_columns:

    if col not in df.columns:

        raise ValueError(
            f"Missing required column: {col}"
        )


# ============================================================
# 6. Metadata
# ============================================================

df["individual"] = (
    df["individual"]
    .astype(str)
)

df["label"] = (
    df["label"]
    .astype(int)
)


print(
    "\nLabel distribution:"
)

print(
    df["label"]
    .value_counts()
)


print(
    "\nDonor × label:"
)

print(
    pd.crosstab(
        df["individual"],
        df["label"]
    )
)


# ============================================================
# 7. Define genes
# ============================================================

gene_columns = [

    col

    for col in df.columns

    if col not in [
        "individual",
        "label"
    ]
]


X = (
    df[
        gene_columns
    ]
    .astype(float)
)

y = (
    df["label"]
    .values
)

groups = (
    df["individual"]
    .values
)

cells = (
    df.index
    .values
)


print(
    "\nCells:",
    X.shape[0]
)

print(
    "Genes:",
    X.shape[1]
)

print(
    "Donors:",
    len(
        np.unique(groups)
    )
)


# ============================================================
# 8. Basic checks
# ============================================================

n_missing = (
    X
    .isna()
    .sum()
    .sum()
)

print(
    "Missing values:",
    n_missing
)


if n_missing > 0:

    raise ValueError(
        "Expression matrix contains missing values."
    )


n_inf = (
    np.isinf(
        X.to_numpy()
    )
    .sum()
)

print(
    "Infinite values:",
    n_inf
)


if n_inf > 0:

    raise ValueError(
        "Expression matrix contains infinite values."
    )


n_duplicate_genes = (
    pd.Index(
        gene_columns
    )
    .duplicated()
    .sum()
)

print(
    "Duplicated gene columns:",
    n_duplicate_genes
)


zero_variance_genes = (

    X.columns[
        X.var(axis=0) == 0
    ]

    .tolist()

)

print(
    "Zero variance genes:",
    len(zero_variance_genes)
)


# ============================================================
# 9. Donors
# ============================================================

donors = np.sort(
    np.unique(groups)
)


print(
    "\nDonors:"
)

print(
    donors
)


# Every donor should contain both HTS and LTS
for donor in donors:

    donor_y = y[
        groups == donor
    ]

    if len(
        np.unique(donor_y)
    ) < 2:

        raise ValueError(
            f"Donor {donor} does not contain both HTS and LTS."
        )


# ============================================================
# 10. donor × class balanced weights
#
# Prevent donors with many cells from dominating training
# ============================================================

def make_donor_class_weights(
    groups,
    y
):

    temp = pd.DataFrame({

        "individual":
            groups,

        "label":
            y

    })


    counts = (

        temp

        .groupby(
            [
                "individual",
                "label"
            ]
        )

        .size()

    )


    weights = []


    for donor, label in zip(
        groups,
        y
    ):

        n = counts.loc[
            (
                donor,
                label
            )
        ]

        weights.append(
            1.0 / n
        )


    weights = np.array(
        weights
    )


    # Mean weight = 1
    weights = (
        weights /
        weights.mean()
    )


    return weights


# ============================================================
# 11. Build GBM
# ============================================================

def build_gbm_model(
    params
):

    model = GradientBoostingClassifier(

        loss="log_loss",

        n_estimators=
            params[
                "n_estimators"
            ],

        learning_rate=
            params[
                "learning_rate"
            ],

        max_depth=
            params[
                "max_depth"
            ],

        min_samples_leaf=5,

        subsample=1.0,

        random_state=
            RANDOM_STATE

    )


    return model


# ============================================================
# 12. Donor-level expression direction
#
# GBM feature importance has no positive/negative direction.
#
# Therefore separately calculate:
#
# mean(HTS) - mean(LTS)
# for each donor
# ============================================================

def donor_expression_direction(
    X_input,
    y_input,
    groups_input
):

    donor_deltas = []


    for donor in np.unique(
        groups_input
    ):

        donor_mask = (
            groups_input
            ==
            donor
        )


        donor_y = (
            y_input[
                donor_mask
            ]
        )


        if len(
            np.unique(
                donor_y
            )
        ) < 2:

            continue


        X_donor = (

            X_input
            .loc[
                donor_mask
            ]

        )


        mean_hts = (

            X_donor
            .loc[
                donor_y == 1
            ]

            .mean(
                axis=0
            )

        )


        mean_lts = (

            X_donor
            .loc[
                donor_y == 0
            ]

            .mean(
                axis=0
            )

        )


        delta = (
            mean_hts
            -
            mean_lts
        )


        donor_deltas.append(
            delta
        )


    delta_df = pd.DataFrame(
        donor_deltas
    )


    mean_delta = (
        delta_df
        .mean(
            axis=0
        )
    )


    positive_fraction = (

        (
            delta_df > 0
        )

        .mean(
            axis=0
        )

    )


    negative_fraction = (

        (
            delta_df < 0
        )

        .mean(
            axis=0
        )

    )


    direction_consistency = (

        pd.concat(
            [
                positive_fraction,
                negative_fraction
            ],
            axis=1
        )

        .max(
            axis=1
        )

    )


    result = pd.DataFrame({

        "gene":
            X_input.columns,

        "mean_HTS_minus_LTS":
            mean_delta[
                X_input.columns
            ].values,

        "expression_direction_consistency":
            direction_consistency[
                X_input.columns
            ].values

    })


    result[
        "expression_direction"
    ] = np.where(

        result[
            "mean_HTS_minus_LTS"
        ] > 0,

        "HTS_high",

        "LTS_high"

    )


    return result


# ============================================================
# 13. Result containers
# ============================================================

outer_metrics = []

all_predictions = []

all_importances = []

all_top20 = []

all_tuning_results = []


# ============================================================
# 14. Outer Leave-One-Donor-Out
# ============================================================

print(
    "\nNumber of outer folds:",
    len(donors)
)


for fold_number, test_donor in enumerate(
    donors,
    start=1
):

    print("\n")
    print("=" * 80)

    print(
        f"Outer fold "
        f"{fold_number}/"
        f"{len(donors)}"
    )

    print(
        f"Test donor: {test_donor}"
    )

    print("=" * 80)


    # ========================================================
    # Outer development / test
    # ========================================================

    test_mask = (
        groups == test_donor
    )

    dev_mask = (
        ~test_mask
    )


    X_dev = (

        X.loc[
            dev_mask
        ]

        .reset_index(
            drop=True
        )

    )


    y_dev = (
        y[
            dev_mask
        ]
    )


    groups_dev = (
        groups[
            dev_mask
        ]
    )


    X_test = (

        X.loc[
            test_mask
        ]

        .reset_index(
            drop=True
        )

    )


    y_test = (
        y[
            test_mask
        ]
    )


    test_cells = (
        cells[
            test_mask
        ]
    )


    print(
        "Development donors:",
        len(
            np.unique(
                groups_dev
            )
        )
    )

    print(
        "Development cells:",
        len(y_dev)
    )

    print(
        "Test cells:",
        len(y_test)
    )


    # ========================================================
    # 15. Inner 4-fold donor CV
    # ========================================================

    inner_cv = GroupKFold(
        n_splits=4
    )


    parameter_scores = []


    # ========================================================
    # Test GBM parameter combinations
    # ========================================================

    for params in GBM_PARAM_GRID:

        inner_AP = []

        inner_ROC = []


        for (
            inner_train_idx,
            inner_val_idx
        ) in inner_cv.split(

            X_dev,

            y_dev,

            groups_dev

        ):


            # =================================================
            # Inner training
            # =================================================

            X_inner_train = (

                X_dev
                .iloc[
                    inner_train_idx
                ]

            )


            y_inner_train = (

                y_dev[
                    inner_train_idx
                ]

            )


            groups_inner_train = (

                groups_dev[
                    inner_train_idx
                ]

            )


            # =================================================
            # Inner validation
            # =================================================

            X_inner_val = (

                X_dev
                .iloc[
                    inner_val_idx
                ]

            )


            y_inner_val = (

                y_dev[
                    inner_val_idx
                ]

            )


            groups_inner_val = (

                groups_dev[
                    inner_val_idx
                ]

            )


            # =================================================
            # donor × class weights
            # =================================================

            train_weights = (

                make_donor_class_weights(

                    groups_inner_train,

                    y_inner_train

                )

            )


            # =================================================
            # Train GBM
            # =================================================

            model = build_gbm_model(
                params
            )


            model.fit(

                X_inner_train,

                y_inner_train,

                sample_weight=
                    train_weights

            )


            # =================================================
            # Validation probability
            # =================================================

            val_prob = (

                model

                .predict_proba(
                    X_inner_val
                )[:, 1]

            )


            # =================================================
            # Donor-macro evaluation
            # =================================================

            val_donors = np.unique(
                groups_inner_val
            )


            for val_donor in val_donors:


                donor_mask = (
                    groups_inner_val
                    ==
                    val_donor
                )


                donor_y = (
                    y_inner_val[
                        donor_mask
                    ]
                )


                donor_prob = (
                    val_prob[
                        donor_mask
                    ]
                )


                if len(
                    np.unique(
                        donor_y
                    )
                ) < 2:

                    continue


                donor_AP = (

                    average_precision_score(

                        donor_y,

                        donor_prob

                    )

                )


                donor_ROC = (

                    roc_auc_score(

                        donor_y,

                        donor_prob

                    )

                )


                inner_AP.append(
                    donor_AP
                )

                inner_ROC.append(
                    donor_ROC
                )


        # ====================================================
        # Mean validation performance
        # ====================================================

        mean_AP = (
            np.mean(
                inner_AP
            )
        )


        mean_ROC = (
            np.mean(
                inner_ROC
            )
        )


        parameter_scores.append({

            "n_estimators":
                params[
                    "n_estimators"
                ],

            "learning_rate":
                params[
                    "learning_rate"
                ],

            "max_depth":
                params[
                    "max_depth"
                ],

            "AP":
                mean_AP,

            "ROC_AUC":
                mean_ROC

        })


        all_tuning_results.append({

            "outer_test_donor":
                test_donor,

            "n_estimators":
                params[
                    "n_estimators"
                ],

            "learning_rate":
                params[
                    "learning_rate"
                ],

            "max_depth":
                params[
                    "max_depth"
                ],

            "inner_macro_AP":
                mean_AP,

            "inner_macro_ROC_AUC":
                mean_ROC

        })


    # ========================================================
    # 16. Select best parameters
    #
    # Primary = AP
    # Secondary = ROC-AUC
    # ========================================================

    parameter_scores_df = pd.DataFrame(
        parameter_scores
    )


    best_row = (

        parameter_scores_df

        .sort_values(

            [
                "AP",
                "ROC_AUC",
                "max_depth",
                "n_estimators"
            ],

            ascending=[
                False,
                False,
                True,
                True
            ]

        )

        .iloc[0]

    )


    best_params = {

        "n_estimators":

            int(
                best_row[
                    "n_estimators"
                ]
            ),

        "learning_rate":

            float(
                best_row[
                    "learning_rate"
                ]
            ),

        "max_depth":

            int(
                best_row[
                    "max_depth"
                ]
            )

    }


    print(
        "\nBest GBM parameters:"
    )

    print(
        best_params
    )


    print(
        "Inner macro AP:",
        round(
            best_row[
                "AP"
            ],
            4
        )
    )


    print(
        "Inner macro ROC-AUC:",
        round(
            best_row[
                "ROC_AUC"
            ],
            4
        )
    )


    # ========================================================
    # 17. Retrain using all development donors
    # ========================================================

    dev_weights = (

        make_donor_class_weights(

            groups_dev,

            y_dev

        )

    )


    final_model = build_gbm_model(
        best_params
    )


    final_model.fit(

        X_dev,

        y_dev,

        sample_weight=
            dev_weights

    )


    # ========================================================
    # 18. Predict completely held-out donor
    # ========================================================

    test_prob = (

        final_model

        .predict_proba(
            X_test
        )[:, 1]

    )


    test_pred = (

        test_prob >= 0.5

    ).astype(int)


    # ========================================================
    # 19. Outer performance
    # ========================================================

    test_ROC = (

        roc_auc_score(

            y_test,

            test_prob

        )

    )


    test_AP = (

        average_precision_score(

            y_test,

            test_prob

        )

    )


    print(
        "Outer ROC-AUC:",
        round(
            test_ROC,
            4
        )
    )


    print(
        "Outer AP:",
        round(
            test_AP,
            4
        )
    )


    outer_metrics.append({

        "fold":
            fold_number,

        "test_donor":
            test_donor,

        "best_n_estimators":
            best_params[
                "n_estimators"
            ],

        "best_learning_rate":
            best_params[
                "learning_rate"
            ],

        "best_max_depth":
            best_params[
                "max_depth"
            ],

        "n_test_cells":
            len(y_test),

        "n_HTS":
            int(
                np.sum(
                    y_test == 1
                )
            ),

        "n_LTS":
            int(
                np.sum(
                    y_test == 0
                )
            ),

        "ROC_AUC":
            test_ROC,

        "AP":
            test_AP

    })


    # ========================================================
    # 20. Predictions
    # ========================================================

    prediction_df = pd.DataFrame({

        "cell":
            test_cells,

        "individual":
            test_donor,

        "true_label":
            y_test,

        "predicted_probability":
            test_prob,

        "predicted_label":
            test_pred

    })


    all_predictions.append(
        prediction_df
    )


    # ========================================================
    # 21. GBM feature importance
    #
    # IMPORTANT:
    # importance only
    # no direction
    # ========================================================

    importance = (
        final_model
        .feature_importances_
    )


    importance_df = pd.DataFrame({

        "outer_test_donor":
            test_donor,

        "gene":
            gene_columns,

        "importance":
            importance

    })


    # ========================================================
    # 22. Add donor-level expression direction
    #
    # Only development donors are used
    # ========================================================

    direction_df = (

        donor_expression_direction(

            X_dev,

            y_dev,

            groups_dev

        )

    )


    importance_df = (

        importance_df

        .merge(

            direction_df,

            on="gene",

            how="left"

        )

    )


    all_importances.append(
        importance_df
    )


    # ========================================================
    # 23. Top20 features
    # ========================================================

    top20 = (

        importance_df

        .sort_values(

            "importance",

            ascending=False

        )

        .head(
            TOP_N
        )

        .copy()

    )


    top20[
        "rank"
    ] = np.arange(

        1,

        len(top20) + 1

    )


    all_top20.append(
        top20
    )


    print(
        "\nTop 10 GBM genes:"
    )


    print(

        top20[
            [
                "gene",
                "importance",
                "mean_HTS_minus_LTS",
                "expression_direction"
            ]
        ]

        .head(10)

        .to_string(
            index=False
        )

    )


# ============================================================
# 24. Combine all results
# ============================================================

outer_metrics_df = pd.DataFrame(
    outer_metrics
)


predictions_df = pd.concat(

    all_predictions,

    ignore_index=True

)


importances_df = pd.concat(

    all_importances,

    ignore_index=True

)


top20_df = pd.concat(

    all_top20,

    ignore_index=True

)


tuning_df = pd.DataFrame(
    all_tuning_results
)


# ============================================================
# 25. Save raw results
# ============================================================

outer_metrics_df.to_csv(

    os.path.join(
        RESULT_DIR,
        "GBM_outer_LODO_performance.csv"
    ),

    index=False

)


predictions_df.to_csv(

    os.path.join(
        RESULT_DIR,
        "GBM_outer_LODO_predictions.csv"
    ),

    index=False

)


importances_df.to_csv(

    os.path.join(
        RESULT_DIR,
        "GBM_all_fold_importances.csv"
    ),

    index=False

)


top20_df.to_csv(

    os.path.join(
        RESULT_DIR,
        "GBM_top20_each_fold.csv"
    ),

    index=False

)


tuning_df.to_csv(

    os.path.join(
        RESULT_DIR,
        "GBM_inner_parameter_tuning.csv"
    ),

    index=False

)


# ============================================================
# 26. Count Top20 frequency
# ============================================================

top20_frequency = (

    top20_df

    .groupby(
        "gene"
    )

    .agg(

        top20_count=(
            "outer_test_donor",
            "nunique"
        )

    )

    .reset_index()

)


# ============================================================
# 27. Feature importance summary
# ============================================================

importance_summary = (

    importances_df

    .groupby(
        "gene"
    )

    .agg(

        mean_importance=(
            "importance",
            "mean"
        ),

        median_importance=(
            "importance",
            "median"
        ),

        sd_importance=(
            "importance",
            "std"
        ),

        mean_HTS_minus_LTS=(
            "mean_HTS_minus_LTS",
            "mean"
        ),

        mean_expression_direction_consistency=(
            "expression_direction_consistency",
            "mean"
        )

    )

    .reset_index()

)


# ============================================================
# 28. Merge stability
# ============================================================

feature_summary = (

    importance_summary

    .merge(

        top20_frequency,

        on="gene",

        how="left"

    )

)


feature_summary[
    "top20_count"
] = (

    feature_summary[
        "top20_count"
    ]

    .fillna(0)

    .astype(int)

)


# ============================================================
# 29. Stable GBM features
#
# Top20 in >=6/9 outer folds
# ============================================================

feature_summary[
    "stable_top20"
] = (

    feature_summary[
        "top20_count"
    ] >= 6

)


# ============================================================
# 30. Expression direction
# ============================================================

feature_summary[
    "expression_direction"
] = np.where(

    feature_summary[
        "mean_HTS_minus_LTS"
    ] > 0,

    "HTS_high",

    "LTS_high"

)


# ============================================================
# 31. Final ranking
# ============================================================

feature_summary = (

    feature_summary

    .sort_values(

        [
            "top20_count",
            "mean_importance"
        ],

        ascending=[
            False,
            False
        ]

    )

    .reset_index(
        drop=True
    )

)


feature_summary[
    "final_rank"
] = np.arange(

    1,

    len(feature_summary) + 1

)


# ============================================================
# 32. Save final feature summary
# ============================================================

feature_summary.to_csv(

    os.path.join(

        RESULT_DIR,

        "GBM_feature_stability_summary.csv"

    ),

    index=False

)


# ============================================================
# 33. Print feature ranking
# ============================================================

print("\n")
print("=" * 85)
print("GBM feature stability ranking")
print("=" * 85)


print(

    feature_summary[
        [
            "final_rank",
            "gene",
            "top20_count",
            "mean_importance",
            "median_importance",
            "mean_HTS_minus_LTS",
            "expression_direction",
            "mean_expression_direction_consistency",
            "stable_top20"
        ]
    ]

    .head(30)

    .to_string(
        index=False
    )

)


# ============================================================
# 34. Performance summary
# ============================================================

roc_median = (
    outer_metrics_df[
        "ROC_AUC"
    ]
    .median()
)


roc_q1 = (
    outer_metrics_df[
        "ROC_AUC"
    ]
    .quantile(0.25)
)


roc_q3 = (
    outer_metrics_df[
        "ROC_AUC"
    ]
    .quantile(0.75)
)


ap_median = (
    outer_metrics_df[
        "AP"
    ]
    .median()
)


ap_q1 = (
    outer_metrics_df[
        "AP"
    ]
    .quantile(0.25)
)


ap_q3 = (
    outer_metrics_df[
        "AP"
    ]
    .quantile(0.75)
)


print("\n")
print("=" * 85)
print("Outer Leave-One-Donor-Out Performance")
print("=" * 85)


print(
    outer_metrics_df
    .to_string(
        index=False
    )
)


print(
    "\nROC-AUC median [IQR]:"
)

print(
    f"{roc_median:.3f} "
    f"[{roc_q1:.3f}, {roc_q3:.3f}]"
)


print(
    "\nAP median [IQR]:"
)

print(
    f"{ap_median:.3f} "
    f"[{ap_q1:.3f}, {ap_q3:.3f}]"
)


# ============================================================
# 35. Save performance summary
# ============================================================

performance_summary = pd.DataFrame({

    "metric": [
        "ROC_AUC",
        "AP"
    ],

    "median": [
        roc_median,
        ap_median
    ],

    "Q1": [
        roc_q1,
        ap_q1
    ],

    "Q3": [
        roc_q3,
        ap_q3
    ]

})


performance_summary.to_csv(

    os.path.join(

        RESULT_DIR,

        "GBM_performance_summary.csv"

    ),

    index=False

)


# ============================================================
# 36. Pooled OOF ROC
#
# visualization only
#
# Formal performance:
# donor-level median [IQR]
# ============================================================

y_oof = (
    predictions_df[
        "true_label"
    ]
    .values
)


prob_oof = (
    predictions_df[
        "predicted_probability"
    ]
    .values
)


fpr, tpr, _ = roc_curve(

    y_oof,

    prob_oof

)


pooled_roc = roc_auc_score(

    y_oof,

    prob_oof

)


plt.figure(
    figsize=(
        6,
        6
    )
)


plt.plot(

    fpr,

    tpr,

    linewidth=2,

    label=(
        f"GBM AUC = "
        f"{pooled_roc:.3f}"
    )

)


plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--"

)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "GBM\nLeave-One-Donor-Out ROC"
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "GBM_LODO_ROC.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 37. Pooled Precision-Recall
# ============================================================

precision, recall, _ = (

    precision_recall_curve(

        y_oof,

        prob_oof

    )

)


pooled_ap = (

    average_precision_score(

        y_oof,

        prob_oof

    )

)


plt.figure(
    figsize=(
        6,
        6
    )
)


plt.plot(

    recall,

    precision,

    linewidth=2,

    label=(
        f"GBM AP = "
        f"{pooled_ap:.3f}"
    )

)


plt.xlabel(
    "Recall"
)

plt.ylabel(
    "Precision"
)

plt.title(
    "GBM\nLeave-One-Donor-Out Precision-Recall"
)

plt.legend(
    loc="lower left"
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "GBM_LODO_PR.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 38. Top20 feature stability
# ============================================================

plot_df = (

    feature_summary

    .head(20)

    .copy()

)


plot_df = (

    plot_df

    .sort_values(

        "top20_count",

        ascending=True

    )

)


plt.figure(
    figsize=(
        7,
        8
    )
)


plt.barh(

    plot_df[
        "gene"
    ],

    plot_df[
        "top20_count"
    ]

)


plt.xlabel(
    "Number of outer folds in Top 20"
)

plt.ylabel(
    "Gene"
)

plt.title(
    "GBM Feature Stability"
)

plt.xlim(
    0,
    len(donors)
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "GBM_Top20_feature_stability.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 39. Donor-level performance
# ============================================================

performance_plot = (

    outer_metrics_df

    .sort_values(
        "ROC_AUC"
    )

    .copy()

)


x = np.arange(
    len(performance_plot)
)

width = 0.35


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.bar(

    x - width / 2,

    performance_plot[
        "ROC_AUC"
    ],

    width,

    label="ROC-AUC"

)


plt.bar(

    x + width / 2,

    performance_plot[
        "AP"
    ],

    width,

    label="AP"

)


plt.xticks(

    x,

    performance_plot[
        "test_donor"
    ],

    rotation=45,

    ha="right"

)


plt.ylim(
    0,
    1
)

plt.ylabel(
    "Performance"
)

plt.xlabel(
    "Held-out donor"
)

plt.title(
    "GBM Performance by Held-out Donor"
)

plt.legend()

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "GBM_outer_donor_performance.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 40. Stable genes
# ============================================================

stable_genes = (

    feature_summary[

        feature_summary[
            "stable_top20"
        ]

    ]

)


print("\n")
print("=" * 85)

print(
    "Stable GBM genes "
    "(Top20 in >= 6/9 folds)"
)

print("=" * 85)


if len(stable_genes) == 0:

    print(
        "No genes reached >= 6/9 Top20 stability."
    )

else:

    print(

        stable_genes[
            [
                "gene",
                "top20_count",
                "mean_importance",
                "median_importance",
                "mean_HTS_minus_LTS",
                "expression_direction",
                "mean_expression_direction_consistency"
            ]
        ]

        .to_string(
            index=False
        )

    )


# ============================================================
# 41. Done
# ============================================================

print(
    "\nAnalysis completed."
)

print(
    f"\nFigures saved to: {FIG_DIR}/"
)

print(
    f"Results saved to: {RESULT_DIR}/"
)
```
