```python
# ============================================================
# Linear SVM
# HTS vs LTS
# Nested donor-aware cross-validation
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GroupKFold

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    roc_curve,
    precision_recall_curve
)


# ============================================================
# 1. Parameters
# ============================================================

DATA_FILE = "AD_microglia_HTS_LTS_TWAS_genes_ML.csv"

FIG_DIR = "figures"
RESULT_DIR = "results"

os.makedirs(FIG_DIR, exist_ok=True)
os.makedirs(RESULT_DIR, exist_ok=True)

RANDOM_STATE = 42

TOP_N = 20


# ------------------------------------------------------------
# Linear SVM C grid
#
# C小：
# regularization更强
#
# C大：
# regularization更弱
# ------------------------------------------------------------

C_GRID = [
    0.001,
    0.01,
    0.1,
    1,
    10,
    100
]


# ============================================================
# 2. Read data
# ============================================================

df = pd.read_csv(DATA_FILE)

print("Original data shape:", df.shape)
print(df.head())


# ============================================================
# 3. Handle cell barcode
# ============================================================

first_col = df.columns[0]

if first_col.startswith("Unnamed"):

    df = df.rename(
        columns={
            first_col: "cell"
        }
    )

    df = df.set_index("cell")

elif first_col.lower() in [
    "cell",
    "barcode",
    "cell_barcode"
]:

    df = df.set_index(first_col)

else:

    df.index = df.index.astype(str)
    df.index.name = "cell"


print("\nData after setting cell index:")
print(df.head())


# ============================================================
# 4. Required columns
# ============================================================

required_columns = [
    "individual",
    "label"
]

for col in required_columns:

    if col not in df.columns:

        raise ValueError(
            f"Missing required column: {col}"
        )


# ============================================================
# 5. Metadata
# ============================================================

df["individual"] = (
    df["individual"]
    .astype(str)
)

df["label"] = (
    df["label"]
    .astype(int)
)


print("\nLabel distribution:")
print(
    df["label"]
    .value_counts()
)


print("\nDonor × label:")
print(
    pd.crosstab(
        df["individual"],
        df["label"]
    )
)


# ============================================================
# 6. Define feature genes
# ============================================================

gene_columns = [

    col

    for col in df.columns

    if col not in [
        "individual",
        "label"
    ]
]


X = (
    df[
        gene_columns
    ]
    .astype(float)
)

y = (
    df["label"]
    .values
)

groups = (
    df["individual"]
    .values
)

cells = (
    df.index
    .values
)


print("\nCells:", X.shape[0])
print("Genes:", X.shape[1])
print(
    "Donors:",
    len(np.unique(groups))
)


# ============================================================
# 7. Data checks
# ============================================================

n_missing = (
    X.isna()
    .sum()
    .sum()
)

print(
    "Missing values:",
    n_missing
)

if n_missing > 0:

    raise ValueError(
        "Expression matrix contains missing values."
    )


n_inf = np.isinf(
    X.to_numpy()
).sum()

print(
    "Infinite values:",
    n_inf
)

if n_inf > 0:

    raise ValueError(
        "Expression matrix contains infinite values."
    )


duplicate_genes = (
    pd.Index(
        gene_columns
    )
    .duplicated()
    .sum()
)

print(
    "Duplicated gene columns:",
    duplicate_genes
)


zero_variance_genes = (
    X.columns[
        X.var(axis=0) == 0
    ]
    .tolist()
)

print(
    "Zero variance genes:",
    len(zero_variance_genes)
)


# ============================================================
# 8. Check donors
#
# 每个donor必须同时有HTS和LTS
# ============================================================

donors = np.sort(
    np.unique(groups)
)

print("\nDonors:")
print(donors)


for donor in donors:

    donor_y = y[
        groups == donor
    ]

    if len(
        np.unique(donor_y)
    ) < 2:

        raise ValueError(
            f"Donor {donor} does not contain both HTS and LTS."
        )


# ============================================================
# 9. donor × class balanced weights
#
# 每个 donor × class 的总贡献基本相同
# ============================================================

def make_donor_class_weights(
    groups,
    y
):

    temp = pd.DataFrame({

        "individual":
            groups,

        "label":
            y

    })


    counts = (

        temp
        .groupby(
            [
                "individual",
                "label"
            ]
        )
        .size()

    )


    weights = []


    for donor, label in zip(
        groups,
        y
    ):

        n = counts.loc[
            (
                donor,
                label
            )
        ]

        weights.append(
            1.0 / n
        )


    weights = np.array(
        weights
    )


    # normalize to mean weight = 1
    weights = (
        weights /
        weights.mean()
    )


    return weights


# ============================================================
# 10. Build Linear SVM model
#
# StandardScaler：
# 必须只在training data中fit
#
# LinearSVC：
# 适合提取feature coefficients
#
# dual=False：
# 当前 cells >> genes
# ============================================================

def build_svm_model(
    C
):

    model = Pipeline([

        (
            "scaler",

            StandardScaler()
        ),

        (
            "svm",

            LinearSVC(

                C=C,

                penalty="l2",

                loss="squared_hinge",

                dual=False,

                max_iter=20000,

                tol=1e-4,

                random_state=
                    RANDOM_STATE

            )
        )

    ])


    return model


# ============================================================
# 11. Donor-level expression direction
#
# 这个方向和SVM coefficient不是一个概念
#
# mean HTS - mean LTS
# ============================================================

def donor_expression_direction(
    X_input,
    y_input,
    groups_input
):

    donor_deltas = []


    for donor in np.unique(
        groups_input
    ):

        donor_mask = (
            groups_input
            ==
            donor
        )


        donor_y = y_input[
            donor_mask
        ]


        if len(
            np.unique(donor_y)
        ) < 2:

            continue


        X_donor = (
            X_input
            .loc[
                donor_mask
            ]
        )


        mean_hts = (

            X_donor
            .loc[
                donor_y == 1
            ]
            .mean(axis=0)

        )


        mean_lts = (

            X_donor
            .loc[
                donor_y == 0
            ]
            .mean(axis=0)

        )


        donor_deltas.append(
            mean_hts -
            mean_lts
        )


    delta_df = pd.DataFrame(
        donor_deltas
    )


    mean_delta = (
        delta_df
        .mean(axis=0)
    )


    positive_fraction = (

        (
            delta_df > 0
        )
        .mean(axis=0)

    )


    negative_fraction = (

        (
            delta_df < 0
        )
        .mean(axis=0)

    )


    direction_consistency = (

        pd.concat(
            [
                positive_fraction,
                negative_fraction
            ],
            axis=1
        )
        .max(axis=1)

    )


    result = pd.DataFrame({

        "gene":
            X_input.columns,

        "mean_HTS_minus_LTS":
            mean_delta[
                X_input.columns
            ].values,

        "expression_direction_consistency":
            direction_consistency[
                X_input.columns
            ].values

    })


    result[
        "expression_direction"
    ] = np.where(

        result[
            "mean_HTS_minus_LTS"
        ] > 0,

        "HTS_high",

        "LTS_high"

    )


    return result


# ============================================================
# 12. Result containers
# ============================================================

outer_metrics = []

all_predictions = []

all_coefficients = []

all_top20 = []

all_tuning_results = []


# ============================================================
# 13. Outer Leave-One-Donor-Out CV
# ============================================================

print(
    "\nNumber of outer folds:",
    len(donors)
)


for fold_number, test_donor in enumerate(
    donors,
    start=1
):

    print("\n")
    print("=" * 75)

    print(
        f"Outer fold "
        f"{fold_number}/"
        f"{len(donors)}"
    )

    print(
        f"Test donor: "
        f"{test_donor}"
    )

    print("=" * 75)


    # ========================================================
    # Outer train / test
    # ========================================================

    test_mask = (
        groups == test_donor
    )

    dev_mask = ~test_mask


    X_dev = (

        X.loc[
            dev_mask
        ]

        .reset_index(
            drop=True
        )

    )


    y_dev = (
        y[
            dev_mask
        ]
    )


    groups_dev = (
        groups[
            dev_mask
        ]
    )


    X_test = (

        X.loc[
            test_mask
        ]

        .reset_index(
            drop=True
        )

    )


    y_test = (
        y[
            test_mask
        ]
    )


    test_cells = (
        cells[
            test_mask
        ]
    )


    print(
        "Development donors:",
        len(
            np.unique(
                groups_dev
            )
        )
    )

    print(
        "Development cells:",
        len(y_dev)
    )

    print(
        "Test cells:",
        len(y_test)
    )


    # ========================================================
    # 14. Inner 4-fold donor CV
    # ========================================================

    inner_cv = GroupKFold(
        n_splits=4
    )


    C_scores = []


    for C in C_GRID:

        inner_AP = []

        inner_ROC = []


        # ====================================================
        # Inner CV
        # ====================================================

        for (
            inner_train_idx,
            inner_val_idx
        ) in inner_cv.split(

            X_dev,

            y_dev,

            groups_dev

        ):


            X_inner_train = (

                X_dev
                .iloc[
                    inner_train_idx
                ]

            )


            y_inner_train = (

                y_dev[
                    inner_train_idx
                ]

            )


            groups_inner_train = (

                groups_dev[
                    inner_train_idx
                ]

            )


            X_inner_val = (

                X_dev
                .iloc[
                    inner_val_idx
                ]

            )


            y_inner_val = (

                y_dev[
                    inner_val_idx
                ]

            )


            groups_inner_val = (

                groups_dev[
                    inner_val_idx
                ]

            )


            # =================================================
            # Training weights
            # =================================================

            train_weights = (

                make_donor_class_weights(

                    groups_inner_train,

                    y_inner_train

                )

            )


            # =================================================
            # Train Linear SVM
            # =================================================

            model = build_svm_model(
                C
            )


            model.fit(

                X_inner_train,

                y_inner_train,

                svm__sample_weight=
                    train_weights

            )


            # =================================================
            # Validation score
            #
            # LinearSVC没有predict_proba
            #
            # 使用decision_function
            # =================================================

            val_score = (

                model
                .decision_function(
                    X_inner_val
                )

            )


            # =================================================
            # donor-macro evaluation
            # =================================================

            val_donors = np.unique(
                groups_inner_val
            )


            for val_donor in val_donors:


                donor_mask = (
                    groups_inner_val
                    ==
                    val_donor
                )


                donor_y = (
                    y_inner_val[
                        donor_mask
                    ]
                )


                donor_score = (
                    val_score[
                        donor_mask
                    ]
                )


                if len(
                    np.unique(
                        donor_y
                    )
                ) < 2:

                    continue


                donor_AP = (

                    average_precision_score(

                        donor_y,

                        donor_score

                    )

                )


                donor_ROC = (

                    roc_auc_score(

                        donor_y,

                        donor_score

                    )

                )


                inner_AP.append(
                    donor_AP
                )

                inner_ROC.append(
                    donor_ROC
                )


        # ====================================================
        # Mean donor-level performance
        # ====================================================

        mean_AP = (
            np.mean(
                inner_AP
            )
        )


        mean_ROC = (
            np.mean(
                inner_ROC
            )
        )


        C_scores.append({

            "C":
                C,

            "AP":
                mean_AP,

            "ROC_AUC":
                mean_ROC

        })


        all_tuning_results.append({

            "outer_test_donor":
                test_donor,

            "C":
                C,

            "inner_macro_AP":
                mean_AP,

            "inner_macro_ROC_AUC":
                mean_ROC

        })


    # ========================================================
    # 15. Select best C
    #
    # AP primary
    # ROC-AUC secondary
    #
    # 如果完全相同：
    # 优先较小C
    # ========================================================

    C_scores_df = pd.DataFrame(
        C_scores
    )


    best_row = (

        C_scores_df

        .sort_values(

            [
                "AP",
                "ROC_AUC",
                "C"
            ],

            ascending=[
                False,
                False,
                True
            ]

        )

        .iloc[0]

    )


    best_C = float(
        best_row[
            "C"
        ]
    )


    print(
        "\nBest C:",
        best_C
    )

    print(
        "Inner macro AP:",
        round(
            best_row[
                "AP"
            ],
            4
        )
    )

    print(
        "Inner macro ROC-AUC:",
        round(
            best_row[
                "ROC_AUC"
            ],
            4
        )
    )


    # ========================================================
    # 16. Retrain on all development donors
    # ========================================================

    dev_weights = (

        make_donor_class_weights(

            groups_dev,

            y_dev

        )

    )


    final_model = build_svm_model(
        best_C
    )


    final_model.fit(

        X_dev,

        y_dev,

        svm__sample_weight=
            dev_weights

    )


    # ========================================================
    # 17. Predict completely held-out donor
    # ========================================================

    test_score = (

        final_model

        .decision_function(
            X_test
        )

    )


    test_pred = (

        test_score >= 0

    ).astype(int)


    # ========================================================
    # 18. Outer test performance
    # ========================================================

    test_ROC = (

        roc_auc_score(

            y_test,

            test_score

        )

    )


    test_AP = (

        average_precision_score(

            y_test,

            test_score

        )

    )


    print(
        "Outer ROC-AUC:",
        round(
            test_ROC,
            4
        )
    )


    print(
        "Outer AP:",
        round(
            test_AP,
            4
        )
    )


    outer_metrics.append({

        "fold":
            fold_number,

        "test_donor":
            test_donor,

        "best_C":
            best_C,

        "n_test_cells":
            len(y_test),

        "n_HTS":
            int(
                np.sum(
                    y_test == 1
                )
            ),

        "n_LTS":
            int(
                np.sum(
                    y_test == 0
                )
            ),

        "ROC_AUC":
            test_ROC,

        "AP":
            test_AP

    })


    # ========================================================
    # 19. Save predictions
    # ========================================================

    prediction_df = pd.DataFrame({

        "cell":
            test_cells,

        "individual":
            test_donor,

        "true_label":
            y_test,

        "decision_score":
            test_score,

        "predicted_label":
            test_pred

    })


    all_predictions.append(
        prediction_df
    )


    # ========================================================
    # 20. SVM coefficients
    #
    # StandardScaler之后：
    # abs(coef)可以用于feature ranking
    #
    # coef > 0：
    # model direction toward HTS
    #
    # coef < 0：
    # model direction toward LTS
    # ========================================================

    coef = (

        final_model
        .named_steps[
            "svm"
        ]
        .coef_[0]

    )


    coefficient_df = pd.DataFrame({

        "outer_test_donor":
            test_donor,

        "gene":
            gene_columns,

        "coefficient":
            coef,

        "abs_coefficient":
            np.abs(coef)

    })


    coefficient_df[
        "model_direction"
    ] = np.where(

        coefficient_df[
            "coefficient"
        ] > 0,

        "HTS_positive",

        "LTS_positive"

    )


    # ========================================================
    # 21. Add actual donor-level expression direction
    #
    # 只使用development donors
    # ========================================================

    direction_df = (

        donor_expression_direction(

            X_dev,

            y_dev,

            groups_dev

        )

    )


    coefficient_df = (

        coefficient_df

        .merge(

            direction_df,

            on="gene",

            how="left"

        )

    )


    all_coefficients.append(
        coefficient_df
    )


    # ========================================================
    # 22. Top20 genes
    # ========================================================

    top20 = (

        coefficient_df

        .sort_values(

            "abs_coefficient",

            ascending=False

        )

        .head(
            TOP_N
        )

        .copy()

    )


    top20[
        "rank"
    ] = np.arange(

        1,

        len(top20) + 1

    )


    all_top20.append(
        top20
    )


    print(
        "\nTop 10 SVM genes:"
    )


    print(

        top20[
            [
                "gene",
                "abs_coefficient",
                "coefficient",
                "model_direction",
                "mean_HTS_minus_LTS",
                "expression_direction"
            ]
        ]

        .head(10)

        .to_string(
            index=False
        )

    )


# ============================================================
# 23. Combine results
# ============================================================

outer_metrics_df = pd.DataFrame(
    outer_metrics
)


predictions_df = pd.concat(

    all_predictions,

    ignore_index=True

)


coefficients_df = pd.concat(

    all_coefficients,

    ignore_index=True

)


top20_df = pd.concat(

    all_top20,

    ignore_index=True

)


tuning_df = pd.DataFrame(
    all_tuning_results
)


# ============================================================
# 24. Save raw results
# ============================================================

outer_metrics_df.to_csv(

    os.path.join(
        RESULT_DIR,
        "SVM_outer_LODO_performance.csv"
    ),

    index=False

)


predictions_df.to_csv(

    os.path.join(
        RESULT_DIR,
        "SVM_outer_LODO_predictions.csv"
    ),

    index=False

)


coefficients_df.to_csv(

    os.path.join(
        RESULT_DIR,
        "SVM_all_fold_coefficients.csv"
    ),

    index=False

)


top20_df.to_csv(

    os.path.join(
        RESULT_DIR,
        "SVM_top20_each_fold.csv"
    ),

    index=False

)


tuning_df.to_csv(

    os.path.join(
        RESULT_DIR,
        "SVM_inner_C_tuning.csv"
    ),

    index=False

)


# ============================================================
# 25. Top20 frequency
# ============================================================

top20_frequency = (

    top20_df

    .groupby(
        "gene"
    )

    .agg(

        top20_count=(
            "outer_test_donor",
            "nunique"
        )

    )

    .reset_index()

)


# ============================================================
# 26. Coefficient summary
# ============================================================

coef_summary = (

    coefficients_df

    .groupby(
        "gene"
    )

    .agg(

        mean_coefficient=(
            "coefficient",
            "mean"
        ),

        mean_abs_coefficient=(
            "abs_coefficient",
            "mean"
        ),

        median_abs_coefficient=(
            "abs_coefficient",
            "median"
        ),

        sd_coefficient=(
            "coefficient",
            "std"
        ),

        positive_fold_count=(
            "coefficient",
            lambda x:
                np.sum(
                    x > 0
                )
        ),

        negative_fold_count=(
            "coefficient",
            lambda x:
                np.sum(
                    x < 0
                )
        ),

        mean_HTS_minus_LTS=(
            "mean_HTS_minus_LTS",
            "mean"
        ),

        mean_expression_direction_consistency=(
            "expression_direction_consistency",
            "mean"
        )

    )

    .reset_index()

)


# ============================================================
# 27. Merge stability
# ============================================================

feature_summary = (

    coef_summary

    .merge(

        top20_frequency,

        on="gene",

        how="left"

    )

)


feature_summary[
    "top20_count"
] = (

    feature_summary[
        "top20_count"
    ]

    .fillna(0)

    .astype(int)

)


# ============================================================
# 28. Stable SVM feature
#
# Top20 in >=6 / 9 outer folds
# ============================================================

feature_summary[
    "stable_top20"
] = (

    feature_summary[
        "top20_count"
    ] >= 6

)


# ============================================================
# 29. SVM model direction
# ============================================================

feature_summary[
    "mean_model_direction"
] = np.where(

    feature_summary[
        "mean_coefficient"
    ] > 0,

    "HTS_positive",

    "LTS_positive"

)


# ============================================================
# 30. SVM coefficient direction consistency
# ============================================================

feature_summary[
    "model_direction_consistency"
] = (

    feature_summary[
        [
            "positive_fold_count",
            "negative_fold_count"
        ]
    ]

    .max(axis=1)

    /

    len(donors)

)


# ============================================================
# 31. Actual expression direction
# ============================================================

feature_summary[
    "expression_direction"
] = np.where(

    feature_summary[
        "mean_HTS_minus_LTS"
    ] > 0,

    "HTS_high",

    "LTS_high"

)


# ============================================================
# 32. Sort feature ranking
# ============================================================

feature_summary = (

    feature_summary

    .sort_values(

        [
            "top20_count",
            "mean_abs_coefficient"
        ],

        ascending=[
            False,
            False
        ]

    )

    .reset_index(
        drop=True
    )

)


feature_summary[
    "final_rank"
] = np.arange(

    1,

    len(feature_summary) + 1

)


# ============================================================
# 33. Save final feature summary
# ============================================================

feature_summary.to_csv(

    os.path.join(

        RESULT_DIR,

        "SVM_feature_stability_summary.csv"

    ),

    index=False

)


# ============================================================
# 34. Print top features
# ============================================================

print("\n")
print("=" * 85)
print("Linear SVM feature stability ranking")
print("=" * 85)


print(

    feature_summary[
        [
            "final_rank",
            "gene",
            "top20_count",
            "mean_abs_coefficient",
            "mean_coefficient",
            "mean_model_direction",
            "model_direction_consistency",
            "mean_HTS_minus_LTS",
            "expression_direction",
            "mean_expression_direction_consistency",
            "stable_top20"
        ]
    ]

    .head(30)

    .to_string(
        index=False
    )

)


# ============================================================
# 35. Performance summary
# ============================================================

roc_median = (
    outer_metrics_df[
        "ROC_AUC"
    ]
    .median()
)

roc_q1 = (
    outer_metrics_df[
        "ROC_AUC"
    ]
    .quantile(0.25)
)

roc_q3 = (
    outer_metrics_df[
        "ROC_AUC"
    ]
    .quantile(0.75)
)


ap_median = (
    outer_metrics_df[
        "AP"
    ]
    .median()
)

ap_q1 = (
    outer_metrics_df[
        "AP"
    ]
    .quantile(0.25)
)

ap_q3 = (
    outer_metrics_df[
        "AP"
    ]
    .quantile(0.75)
)


print("\n")
print("=" * 85)
print("Outer Leave-One-Donor-Out Performance")
print("=" * 85)


print(
    outer_metrics_df
    .to_string(
        index=False
    )
)


print("\nROC-AUC median [IQR]:")

print(
    f"{roc_median:.3f} "
    f"[{roc_q1:.3f}, {roc_q3:.3f}]"
)


print("\nAP median [IQR]:")

print(
    f"{ap_median:.3f} "
    f"[{ap_q1:.3f}, {ap_q3:.3f}]"
)


# ============================================================
# 36. Save performance summary
# ============================================================

performance_summary = pd.DataFrame({

    "metric": [
        "ROC_AUC",
        "AP"
    ],

    "median": [
        roc_median,
        ap_median
    ],

    "Q1": [
        roc_q1,
        ap_q1
    ],

    "Q3": [
        roc_q3,
        ap_q3
    ]

})


performance_summary.to_csv(

    os.path.join(

        RESULT_DIR,

        "SVM_performance_summary.csv"

    ),

    index=False

)


# ============================================================
# 37. Pooled OOF ROC
#
# visualization only
#
# 正式报告优先使用：
# donor-level median [IQR]
# ============================================================

y_oof = (
    predictions_df[
        "true_label"
    ]
    .values
)


score_oof = (
    predictions_df[
        "decision_score"
    ]
    .values
)


fpr, tpr, _ = roc_curve(

    y_oof,

    score_oof

)


pooled_roc = roc_auc_score(

    y_oof,

    score_oof

)


plt.figure(
    figsize=(
        6,
        6
    )
)


plt.plot(

    fpr,

    tpr,

    linewidth=2,

    label=(
        f"Linear SVM AUC = "
        f"{pooled_roc:.3f}"
    )

)


plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--"

)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "Linear SVM\nLeave-One-Donor-Out ROC"
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "SVM_LODO_ROC.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 38. Pooled OOF Precision-Recall
# ============================================================

precision, recall, _ = (

    precision_recall_curve(

        y_oof,

        score_oof

    )

)


pooled_ap = (

    average_precision_score(

        y_oof,

        score_oof

    )

)


plt.figure(
    figsize=(
        6,
        6
    )
)


plt.plot(

    recall,

    precision,

    linewidth=2,

    label=(
        f"Linear SVM AP = "
        f"{pooled_ap:.3f}"
    )

)


plt.xlabel(
    "Recall"
)

plt.ylabel(
    "Precision"
)

plt.title(
    "Linear SVM\n"
    "Leave-One-Donor-Out Precision-Recall"
)

plt.legend(
    loc="lower left"
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "SVM_LODO_PR.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 39. Feature stability figure
# ============================================================

plot_df = (

    feature_summary

    .head(20)

    .copy()

)


plot_df = (

    plot_df

    .sort_values(

        "top20_count",

        ascending=True

    )

)


plt.figure(
    figsize=(
        7,
        8
    )
)


plt.barh(

    plot_df[
        "gene"
    ],

    plot_df[
        "top20_count"
    ]

)


plt.xlabel(
    "Number of outer folds in Top 20"
)

plt.ylabel(
    "Gene"
)

plt.title(
    "Linear SVM Feature Stability"
)

plt.xlim(
    0,
    len(donors)
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "SVM_Top20_feature_stability.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 40. Donor-level performance plot
# ============================================================

performance_plot = (

    outer_metrics_df

    .sort_values(
        "ROC_AUC"
    )

    .copy()

)


x = np.arange(
    len(performance_plot)
)

width = 0.35


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.bar(

    x - width / 2,

    performance_plot[
        "ROC_AUC"
    ],

    width,

    label="ROC-AUC"

)


plt.bar(

    x + width / 2,

    performance_plot[
        "AP"
    ],

    width,

    label="AP"

)


plt.xticks(

    x,

    performance_plot[
        "test_donor"
    ],

    rotation=45,

    ha="right"

)


plt.ylim(
    0,
    1
)

plt.ylabel(
    "Performance"
)

plt.xlabel(
    "Held-out donor"
)

plt.title(
    "Linear SVM Performance by Held-out Donor"
)

plt.legend()

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "SVM_outer_donor_performance.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 41. Print stable genes
# ============================================================

stable_genes = (

    feature_summary[

        feature_summary[
            "stable_top20"
        ]

    ]

)


print("\n")
print("=" * 85)
print("Stable Linear SVM genes (Top20 in >= 6/9 folds)")
print("=" * 85)


if len(stable_genes) == 0:

    print(
        "No genes reached >= 6/9 Top20 stability."
    )

else:

    print(

        stable_genes[
            [
                "gene",
                "top20_count",
                "mean_abs_coefficient",
                "mean_model_direction",
                "model_direction_consistency",
                "mean_HTS_minus_LTS",
                "expression_direction",
                "mean_expression_direction_consistency"
            ]
        ]

        .to_string(
            index=False
        )

    )


# ============================================================
# 42. Done
# ============================================================

print("\nAnalysis completed.")

print(
    f"\nFigures saved to: "
    f"{FIG_DIR}/"
)

print(
    f"Results saved to: "
    f"{RESULT_DIR}/"
)
```
