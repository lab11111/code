```python
# ============================================================
# XGBoost
# HTS vs LTS
# Nested donor-aware cross-validation
#
# Feature ranking:
# donor-balanced mean(|SHAP|)
#
# HTS = 1
# LTS = 0
# ============================================================


# ============================================================
# 0. Import
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from xgboost import XGBClassifier
import shap

from sklearn.model_selection import GroupKFold

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    roc_curve,
    precision_recall_curve
)


# ============================================================
# 1. Parameters
# ============================================================

DATA_FILE = "AD_microglia_HTS_LTS_TWAS_genes_ML.csv"

FIG_DIR = "figures"
RESULT_DIR = "results"

os.makedirs(
    FIG_DIR,
    exist_ok=True
)

os.makedirs(
    RESULT_DIR,
    exist_ok=True
)


RANDOM_STATE = 42

TOP_N = 20


# ============================================================
# 2. XGBoost parameter grid
#
# 2 × 2 × 2 = 8 combinations
#
# 这样nested CV计算量不会过大
# ============================================================

XGB_PARAM_GRID = []


for n_estimators in [
    100,
    200
]:

    for learning_rate in [
        0.03,
        0.1
    ]:

        for max_depth in [
            2,
            3
        ]:

            XGB_PARAM_GRID.append({

                "n_estimators":
                    n_estimators,

                "learning_rate":
                    learning_rate,

                "max_depth":
                    max_depth

            })


print(
    "Number of XGBoost parameter combinations:",
    len(XGB_PARAM_GRID)
)


# ============================================================
# 3. Read data
# ============================================================

df = pd.read_csv(
    DATA_FILE
)


print(
    "Original data shape:",
    df.shape
)

print(
    df.head()
)


# ============================================================
# 4. Cell barcode
# ============================================================

first_col = df.columns[0]


if first_col.startswith(
    "Unnamed"
):

    df = df.rename(

        columns={
            first_col:
                "cell"
        }

    )

    df = df.set_index(
        "cell"
    )


elif first_col.lower() in [

    "cell",
    "barcode",
    "cell_barcode"

]:

    df = df.set_index(
        first_col
    )


else:

    df.index = (
        df.index
        .astype(str)
    )

    df.index.name = "cell"


print(
    "\nData after setting cell index:"
)

print(
    df.head()
)


# ============================================================
# 5. Required columns
# ============================================================

required_columns = [

    "individual",
    "label"

]


for col in required_columns:

    if col not in df.columns:

        raise ValueError(
            f"Missing required column: {col}"
        )


# ============================================================
# 6. Metadata
# ============================================================

df[
    "individual"
] = (

    df[
        "individual"
    ]

    .astype(str)

)


df[
    "label"
] = (

    df[
        "label"
    ]

    .astype(int)

)


print(
    "\nLabel distribution:"
)

print(
    df[
        "label"
    ]
    .value_counts()
)


print(
    "\nDonor × label:"
)

print(

    pd.crosstab(

        df[
            "individual"
        ],

        df[
            "label"
        ]

    )

)


# ============================================================
# 7. Feature genes
# ============================================================

gene_columns = [

    col

    for col in df.columns

    if col not in [

        "individual",
        "label"

    ]

]


X = (

    df[
        gene_columns
    ]

    .astype(float)

)


y = (

    df[
        "label"
    ]

    .values

)


groups = (

    df[
        "individual"
    ]

    .values

)


cells = (
    df.index.values
)


print(
    "\nCells:",
    X.shape[0]
)

print(
    "Genes:",
    X.shape[1]
)

print(
    "Donors:",
    len(
        np.unique(
            groups
        )
    )
)


# ============================================================
# 8. Basic checks
# ============================================================

n_missing = (

    X
    .isna()
    .sum()
    .sum()

)


print(
    "Missing values:",
    n_missing
)


if n_missing > 0:

    raise ValueError(

        "Expression matrix contains missing values."

    )


# ------------------------------------------------------------

n_inf = (

    np.isinf(
        X.to_numpy()
    )

    .sum()

)


print(
    "Infinite values:",
    n_inf
)


if n_inf > 0:

    raise ValueError(

        "Expression matrix contains infinite values."

    )


# ------------------------------------------------------------

n_duplicate_genes = (

    pd.Index(
        gene_columns
    )

    .duplicated()
    .sum()

)


print(
    "Duplicated gene columns:",
    n_duplicate_genes
)


# ------------------------------------------------------------

zero_variance_genes = (

    X.columns[
        X.var(
            axis=0
        ) == 0
    ]

    .tolist()

)


print(
    "Zero variance genes:",
    len(
        zero_variance_genes
    )
)


# ============================================================
# 9. Donors
# ============================================================

donors = np.sort(

    np.unique(
        groups
    )

)


print(
    "\nDonors:"
)

print(
    donors
)


# 每个donor必须同时存在HTS和LTS
for donor in donors:

    donor_y = y[
        groups == donor
    ]

    if len(
        np.unique(
            donor_y
        )
    ) < 2:

        raise ValueError(

            f"Donor {donor} "
            "does not contain both HTS and LTS."

        )


# ============================================================
# 10. donor × class balanced weights
#
# 例如：
#
# donor A × HTS
# donor A × LTS
# donor B × HTS
# donor B × LTS
#
# 每一个组合总权重近似相等
# ============================================================

def make_donor_class_weights(
    groups,
    y
):

    temp = pd.DataFrame({

        "individual":
            groups,

        "label":
            y

    })


    counts = (

        temp

        .groupby(
            [
                "individual",
                "label"
            ]
        )

        .size()

    )


    weights = []


    for donor, label in zip(
        groups,
        y
    ):

        n = counts.loc[
            (
                donor,
                label
            )
        ]

        weights.append(
            1.0 / n
        )


    weights = np.array(
        weights
    )


    # 平均权重标准化到1
    weights = (

        weights /
        weights.mean()

    )


    return weights


# ============================================================
# 11. Build XGBoost
#
# tree_method = hist
# CPU运行
#
# 不需要StandardScaler
# ============================================================

def build_xgb_model(
    params
):

    model = XGBClassifier(

        objective=
            "binary:logistic",

        eval_metric=
            "logloss",

        n_estimators=
            params[
                "n_estimators"
            ],

        learning_rate=
            params[
                "learning_rate"
            ],

        max_depth=
            params[
                "max_depth"
            ],

        min_child_weight=
            1,

        subsample=
            0.8,

        colsample_bytree=
            0.8,

        reg_alpha=
            0.0,

        reg_lambda=
            1.0,

        tree_method=
            "hist",

        random_state=
            RANDOM_STATE,

        n_jobs=
            -1

    )


    return model


# ============================================================
# 12. Donor-level actual expression direction
#
# 注意：
#
# 这个和SHAP方向不是同一个概念。
#
# 对每个development donor计算：
#
# mean expression HTS
# -
# mean expression LTS
# ============================================================

def donor_expression_direction(
    X_input,
    y_input,
    groups_input
):

    donor_deltas = []


    for donor in np.unique(
        groups_input
    ):

        donor_mask = (

            groups_input
            ==
            donor

        )


        donor_y = (

            y_input[
                donor_mask
            ]

        )


        if len(
            np.unique(
                donor_y
            )
        ) < 2:

            continue


        X_donor = (

            X_input.loc[
                donor_mask
            ]

        )


        mean_hts = (

            X_donor.loc[
                donor_y == 1
            ]

            .mean(
                axis=0
            )

        )


        mean_lts = (

            X_donor.loc[
                donor_y == 0
            ]

            .mean(
                axis=0
            )

        )


        donor_deltas.append(

            mean_hts
            -
            mean_lts

        )


    delta_df = pd.DataFrame(
        donor_deltas
    )


    mean_delta = (

        delta_df
        .mean(
            axis=0
        )

    )


    positive_fraction = (

        (
            delta_df > 0
        )

        .mean(
            axis=0
        )

    )


    negative_fraction = (

        (
            delta_df < 0
        )

        .mean(
            axis=0
        )

    )


    direction_consistency = (

        pd.concat(

            [
                positive_fraction,
                negative_fraction
            ],

            axis=1

        )

        .max(
            axis=1
        )

    )


    result = pd.DataFrame({

        "gene":
            X_input.columns,

        "mean_HTS_minus_LTS":

            mean_delta[
                X_input.columns
            ].values,

        "expression_direction_consistency":

            direction_consistency[
                X_input.columns
            ].values

    })


    result[
        "expression_direction"
    ] = np.where(

        result[
            "mean_HTS_minus_LTS"
        ] > 0,

        "HTS_high",

        "LTS_high"

    )


    return result


# ============================================================
# 13. Donor-balanced SHAP importance
#
# 关键步骤
#
# 不能直接：
#
# mean(abs(SHAP)) across all cells
#
# 因为cell多的donor会贡献更多。
#
# 正确：
#
# donor 1 → mean(|SHAP|)
# donor 2 → mean(|SHAP|)
# ...
#
# 最后8个development donors等权平均。
# ============================================================

def donor_balanced_shap_importance(

    model,

    X_input,

    groups_input

):

    # --------------------------------------------------------
    # TreeExplainer
    # --------------------------------------------------------

    explainer = shap.TreeExplainer(

        model,

        feature_perturbation=
            "tree_path_dependent",

        model_output=
            "raw"

    )


    shap_values = (

        explainer
        .shap_values(

            X_input,

            check_additivity=False

        )

    )


    # ========================================================
    # Compatibility across SHAP versions
    # ========================================================

    if isinstance(
        shap_values,
        list
    ):

        # 二分类某些旧版本可能返回list
        shap_values = (
            shap_values[-1]
        )


    shap_values = np.asarray(
        shap_values
    )


    # --------------------------------------------------------
    # 某些版本可能出现：
    #
    # cells × genes × classes
    # --------------------------------------------------------

    if shap_values.ndim == 3:

        if shap_values.shape[-1] == 2:

            shap_values = (
                shap_values[
                    :,
                    :,
                    1
                ]
            )

        elif shap_values.shape[0] == 2:

            shap_values = (
                shap_values[1]
            )


    # --------------------------------------------------------
    # Final check
    # --------------------------------------------------------

    if shap_values.shape[0] != X_input.shape[0]:

        raise ValueError(
            "Unexpected SHAP number of cells."
        )


    if shap_values.shape[1] != X_input.shape[1]:

        raise ValueError(
            "Unexpected SHAP number of genes."
        )


    # ========================================================
    # donor-wise SHAP
    # ========================================================

    donor_abs_shap = []

    donor_signed_shap = []


    donor_names = np.unique(
        groups_input
    )


    for donor in donor_names:

        donor_mask = (

            groups_input
            ==
            donor

        )


        donor_shap = (

            shap_values[
                donor_mask,
                :
            ]

        )


        # ----------------------------------------------------
        # mean absolute SHAP per gene
        # ----------------------------------------------------

        donor_abs = (

            np.mean(

                np.abs(
                    donor_shap
                ),

                axis=0

            )

        )


        # ----------------------------------------------------
        # signed SHAP
        #
        # 仅作为模型贡献参考
        # 不作为基因真实表达方向
        # ----------------------------------------------------

        donor_signed = (

            np.mean(

                donor_shap,

                axis=0

            )

        )


        donor_abs_shap.append(
            donor_abs
        )

        donor_signed_shap.append(
            donor_signed
        )


    donor_abs_shap = np.vstack(
        donor_abs_shap
    )


    donor_signed_shap = np.vstack(
        donor_signed_shap
    )


    # ========================================================
    # equal-weight across donors
    # ========================================================

    mean_abs_shap = (

        donor_abs_shap
        .mean(
            axis=0
        )

    )


    median_abs_shap = (

        np.median(

            donor_abs_shap,

            axis=0

        )

    )


    mean_signed_shap = (

        donor_signed_shap
        .mean(
            axis=0
        )

    )


    # ========================================================
    # SHAP importance donor consistency
    #
    # CV = SD / mean
    #
    # 小 = donor间importance更稳定
    # ========================================================

    sd_abs_shap = (

        donor_abs_shap
        .std(
            axis=0,
            ddof=1
        )

    )


    result = pd.DataFrame({

        "gene":
            X_input.columns,

        "mean_abs_SHAP":
            mean_abs_shap,

        "median_abs_SHAP":
            median_abs_shap,

        "sd_abs_SHAP":
            sd_abs_shap,

        "mean_signed_SHAP":
            mean_signed_shap

    })


    return result


# ============================================================
# 14. Result containers
# ============================================================

outer_metrics = []

all_predictions = []

all_shap_importance = []

all_top20 = []

all_tuning_results = []


# ============================================================
# 15. Outer Leave-One-Donor-Out
# ============================================================

print(
    "\nNumber of outer folds:",
    len(donors)
)


for fold_number, test_donor in enumerate(

    donors,

    start=1

):


    print("\n")

    print(
        "=" * 85
    )


    print(

        f"Outer fold "
        f"{fold_number}/"
        f"{len(donors)}"

    )


    print(

        f"Test donor: "
        f"{test_donor}"

    )


    print(
        "=" * 85
    )


    # ========================================================
    # Outer development/test split
    # ========================================================

    test_mask = (

        groups
        ==
        test_donor

    )


    dev_mask = (
        ~test_mask
    )


    X_dev = (

        X.loc[
            dev_mask
        ]

        .reset_index(
            drop=True
        )

    )


    y_dev = (

        y[
            dev_mask
        ]

    )


    groups_dev = (

        groups[
            dev_mask
        ]

    )


    X_test = (

        X.loc[
            test_mask
        ]

        .reset_index(
            drop=True
        )

    )


    y_test = (

        y[
            test_mask
        ]

    )


    test_cells = (

        cells[
            test_mask
        ]

    )


    print(

        "Development donors:",

        len(
            np.unique(
                groups_dev
            )
        )

    )


    print(

        "Development cells:",

        len(
            y_dev
        )

    )


    print(

        "Test cells:",

        len(
            y_test
        )

    )


    # ========================================================
    # 16. Inner 4-fold donor CV
    # ========================================================

    inner_cv = GroupKFold(

        n_splits=4

    )


    parameter_scores = []


    # ========================================================
    # Parameter tuning
    # ========================================================

    for params in XGB_PARAM_GRID:


        inner_AP = []

        inner_ROC = []


        # ----------------------------------------------------
        # Inner donor folds
        # ----------------------------------------------------

        for (

            inner_train_idx,
            inner_val_idx

        ) in inner_cv.split(

            X_dev,

            y_dev,

            groups_dev

        ):


            X_inner_train = (

                X_dev.iloc[
                    inner_train_idx
                ]

            )


            y_inner_train = (

                y_dev[
                    inner_train_idx
                ]

            )


            groups_inner_train = (

                groups_dev[
                    inner_train_idx
                ]

            )


            X_inner_val = (

                X_dev.iloc[
                    inner_val_idx
                ]

            )


            y_inner_val = (

                y_dev[
                    inner_val_idx
                ]

            )


            groups_inner_val = (

                groups_dev[
                    inner_val_idx
                ]

            )


            # =================================================
            # donor × class weights
            # =================================================

            train_weights = (

                make_donor_class_weights(

                    groups_inner_train,

                    y_inner_train

                )

            )


            # =================================================
            # Train XGBoost
            # =================================================

            model = build_xgb_model(
                params
            )


            model.fit(

                X_inner_train,

                y_inner_train,

                sample_weight=
                    train_weights,

                verbose=False

            )


            # =================================================
            # Validation prediction
            # =================================================

            val_prob = (

                model

                .predict_proba(
                    X_inner_val
                )[:, 1]

            )


            # =================================================
            # donor-macro evaluation
            # =================================================

            val_donors = np.unique(
                groups_inner_val
            )


            for val_donor in val_donors:


                donor_mask = (

                    groups_inner_val
                    ==
                    val_donor

                )


                donor_y = (

                    y_inner_val[
                        donor_mask
                    ]

                )


                donor_prob = (

                    val_prob[
                        donor_mask
                    ]

                )


                if len(
                    np.unique(
                        donor_y
                    )
                ) < 2:

                    continue


                donor_AP = (

                    average_precision_score(

                        donor_y,

                        donor_prob

                    )

                )


                donor_ROC = (

                    roc_auc_score(

                        donor_y,

                        donor_prob

                    )

                )


                inner_AP.append(
                    donor_AP
                )


                inner_ROC.append(
                    donor_ROC
                )


        # ====================================================
        # Mean donor performance
        # ====================================================

        mean_AP = (

            np.mean(
                inner_AP
            )

        )


        mean_ROC = (

            np.mean(
                inner_ROC
            )

        )


        parameter_scores.append({

            "n_estimators":
                params[
                    "n_estimators"
                ],

            "learning_rate":
                params[
                    "learning_rate"
                ],

            "max_depth":
                params[
                    "max_depth"
                ],

            "AP":
                mean_AP,

            "ROC_AUC":
                mean_ROC

        })


        all_tuning_results.append({

            "outer_test_donor":
                test_donor,

            "n_estimators":
                params[
                    "n_estimators"
                ],

            "learning_rate":
                params[
                    "learning_rate"
                ],

            "max_depth":
                params[
                    "max_depth"
                ],

            "inner_macro_AP":
                mean_AP,

            "inner_macro_ROC_AUC":
                mean_ROC

        })


    # ========================================================
    # 17. Select best parameters
    #
    # Primary:
    # AP
    #
    # Secondary:
    # ROC-AUC
    # ========================================================

    parameter_scores_df = pd.DataFrame(
        parameter_scores
    )


    best_row = (

        parameter_scores_df

        .sort_values(

            [
                "AP",
                "ROC_AUC",
                "max_depth",
                "n_estimators"
            ],

            ascending=[
                False,
                False,
                True,
                True
            ]

        )

        .iloc[0]

    )


    best_params = {

        "n_estimators":

            int(
                best_row[
                    "n_estimators"
                ]
            ),

        "learning_rate":

            float(
                best_row[
                    "learning_rate"
                ]
            ),

        "max_depth":

            int(
                best_row[
                    "max_depth"
                ]
            )

    }


    print(
        "\nBest XGBoost parameters:"
    )

    print(
        best_params
    )


    print(

        "Inner macro AP:",

        round(
            best_row[
                "AP"
            ],
            4
        )

    )


    print(

        "Inner macro ROC-AUC:",

        round(
            best_row[
                "ROC_AUC"
            ],
            4
        )

    )


    # ========================================================
    # 18. Retrain on all 8 development donors
    # ========================================================

    dev_weights = (

        make_donor_class_weights(

            groups_dev,

            y_dev

        )

    )


    final_model = build_xgb_model(
        best_params
    )


    final_model.fit(

        X_dev,

        y_dev,

        sample_weight=
            dev_weights,

        verbose=False

    )


    # ========================================================
    # 19. Outer test donor prediction
    #
    # 第一次使用test donor
    # ========================================================

    test_prob = (

        final_model

        .predict_proba(
            X_test
        )[:, 1]

    )


    test_pred = (

        test_prob
        >= 0.5

    ).astype(
        int
    )


    # ========================================================
    # 20. Outer performance
    # ========================================================

    test_ROC = (

        roc_auc_score(

            y_test,

            test_prob

        )

    )


    test_AP = (

        average_precision_score(

            y_test,

            test_prob

        )

    )


    print(

        "Outer ROC-AUC:",

        round(
            test_ROC,
            4
        )

    )


    print(

        "Outer AP:",

        round(
            test_AP,
            4
        )

    )


    outer_metrics.append({

        "fold":
            fold_number,

        "test_donor":
            test_donor,

        "best_n_estimators":
            best_params[
                "n_estimators"
            ],

        "best_learning_rate":
            best_params[
                "learning_rate"
            ],

        "best_max_depth":
            best_params[
                "max_depth"
            ],

        "n_test_cells":
            len(
                y_test
            ),

        "n_HTS":
            int(
                np.sum(
                    y_test == 1
                )
            ),

        "n_LTS":
            int(
                np.sum(
                    y_test == 0
                )
            ),

        "ROC_AUC":
            test_ROC,

        "AP":
            test_AP

    })


    # ========================================================
    # 21. Save predictions
    # ========================================================

    prediction_df = pd.DataFrame({

        "cell":
            test_cells,

        "individual":
            test_donor,

        "true_label":
            y_test,

        "predicted_probability":
            test_prob,

        "predicted_label":
            test_pred

    })


    all_predictions.append(
        prediction_df
    )


    # ========================================================
    # 22. SHAP feature importance
    #
    # IMPORTANT:
    #
    # SHAP只在development donors计算
    #
    # Outer test donor完全不参与feature ranking
    # ========================================================

    print(
        "Calculating donor-balanced SHAP..."
    )


    shap_df = (

        donor_balanced_shap_importance(

            final_model,

            X_dev,

            groups_dev

        )

    )


    shap_df[
        "outer_test_donor"
    ] = test_donor


    # ========================================================
    # 23. Add expression direction
    # ========================================================

    direction_df = (

        donor_expression_direction(

            X_dev,

            y_dev,

            groups_dev

        )

    )


    shap_df = (

        shap_df

        .merge(

            direction_df,

            on="gene",

            how="left"

        )

    )


    all_shap_importance.append(
        shap_df
    )


    # ========================================================
    # 24. Top20 genes
    #
    # 排名依据：
    #
    # donor-balanced mean(|SHAP|)
    # ========================================================

    top20 = (

        shap_df

        .sort_values(

            "mean_abs_SHAP",

            ascending=False

        )

        .head(
            TOP_N
        )

        .copy()

    )


    top20[
        "rank"
    ] = np.arange(

        1,

        len(
            top20
        ) + 1

    )


    all_top20.append(
        top20
    )


    print(
        "\nTop 10 XGBoost genes:"
    )


    print(

        top20[
            [
                "gene",
                "mean_abs_SHAP",
                "median_abs_SHAP",
                "mean_HTS_minus_LTS",
                "expression_direction"
            ]
        ]

        .head(10)

        .to_string(
            index=False
        )

    )


# ============================================================
# 25. Combine all results
# ============================================================

outer_metrics_df = pd.DataFrame(
    outer_metrics
)


predictions_df = pd.concat(

    all_predictions,

    ignore_index=True

)


shap_importance_df = pd.concat(

    all_shap_importance,

    ignore_index=True

)


top20_df = pd.concat(

    all_top20,

    ignore_index=True

)


tuning_df = pd.DataFrame(
    all_tuning_results
)


# ============================================================
# 26. Save raw results
# ============================================================

outer_metrics_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "XGB_outer_LODO_performance.csv"

    ),

    index=False

)


predictions_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "XGB_outer_LODO_predictions.csv"

    ),

    index=False

)


shap_importance_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "XGB_all_fold_SHAP_importance.csv"

    ),

    index=False

)


top20_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "XGB_top20_each_fold.csv"

    ),

    index=False

)


tuning_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "XGB_inner_parameter_tuning.csv"

    ),

    index=False

)


# ============================================================
# 27. Top20 frequency
# ============================================================

top20_frequency = (

    top20_df

    .groupby(
        "gene"
    )

    .agg(

        top20_count=(
            "outer_test_donor",
            "nunique"
        )

    )

    .reset_index()

)


# ============================================================
# 28. SHAP summary across outer folds
# ============================================================

shap_summary = (

    shap_importance_df

    .groupby(
        "gene"
    )

    .agg(

        mean_abs_SHAP=(
            "mean_abs_SHAP",
            "mean"
        ),

        median_abs_SHAP=(
            "mean_abs_SHAP",
            "median"
        ),

        sd_abs_SHAP=(
            "mean_abs_SHAP",
            "std"
        ),

        mean_signed_SHAP=(
            "mean_signed_SHAP",
            "mean"
        ),

        mean_HTS_minus_LTS=(
            "mean_HTS_minus_LTS",
            "mean"
        ),

        mean_expression_direction_consistency=(
            "expression_direction_consistency",
            "mean"
        )

    )

    .reset_index()

)


# ============================================================
# 29. Merge feature stability
# ============================================================

feature_summary = (

    shap_summary

    .merge(

        top20_frequency,

        on="gene",

        how="left"

    )

)


feature_summary[
    "top20_count"
] = (

    feature_summary[
        "top20_count"
    ]

    .fillna(0)

    .astype(int)

)


# ============================================================
# 30. Stable XGBoost feature
#
# Top20 in >= 6 / 9 outer folds
# ============================================================

feature_summary[
    "stable_top20"
] = (

    feature_summary[
        "top20_count"
    ]

    >= 6

)


# ============================================================
# 31. Actual expression direction
# ============================================================

feature_summary[
    "expression_direction"
] = np.where(

    feature_summary[
        "mean_HTS_minus_LTS"
    ] > 0,

    "HTS_high",

    "LTS_high"

)


# ============================================================
# 32. Final ranking
#
# First:
# stability
#
# Second:
# mean absolute SHAP
# ============================================================

feature_summary = (

    feature_summary

    .sort_values(

        [
            "top20_count",
            "mean_abs_SHAP"
        ],

        ascending=[
            False,
            False
        ]

    )

    .reset_index(
        drop=True
    )

)


feature_summary[
    "final_rank"
] = np.arange(

    1,

    len(
        feature_summary
    ) + 1

)


# ============================================================
# 33. Save final feature summary
# ============================================================

feature_summary.to_csv(

    os.path.join(

        RESULT_DIR,

        "XGB_feature_stability_summary.csv"

    ),

    index=False

)


# ============================================================
# 34. Print top genes
# ============================================================

print("\n")

print(
    "=" * 90
)

print(
    "XGBoost feature stability ranking"
)

print(
    "=" * 90
)


print(

    feature_summary[
        [
            "final_rank",
            "gene",
            "top20_count",
            "mean_abs_SHAP",
            "median_abs_SHAP",
            "mean_HTS_minus_LTS",
            "expression_direction",
            "mean_expression_direction_consistency",
            "stable_top20"
        ]
    ]

    .head(30)

    .to_string(
        index=False
    )

)


# ============================================================
# 35. Outer performance summary
# ============================================================

roc_median = (

    outer_metrics_df[
        "ROC_AUC"
    ]

    .median()

)


roc_q1 = (

    outer_metrics_df[
        "ROC_AUC"
    ]

    .quantile(
        0.25
    )

)


roc_q3 = (

    outer_metrics_df[
        "ROC_AUC"
    ]

    .quantile(
        0.75
    )

)


ap_median = (

    outer_metrics_df[
        "AP"
    ]

    .median()

)


ap_q1 = (

    outer_metrics_df[
        "AP"
    ]

    .quantile(
        0.25
    )

)


ap_q3 = (

    outer_metrics_df[
        "AP"
    ]

    .quantile(
        0.75
    )

)


print("\n")

print(
    "=" * 90
)

print(
    "Outer Leave-One-Donor-Out Performance"
)

print(
    "=" * 90
)


print(

    outer_metrics_df

    .to_string(
        index=False
    )

)


print(
    "\nROC-AUC median [IQR]:"
)


print(

    f"{roc_median:.3f} "
    f"[{roc_q1:.3f}, "
    f"{roc_q3:.3f}]"

)


print(
    "\nAP median [IQR]:"
)


print(

    f"{ap_median:.3f} "
    f"[{ap_q1:.3f}, "
    f"{ap_q3:.3f}]"

)


# ============================================================
# 36. Save performance summary
# ============================================================

performance_summary = pd.DataFrame({

    "metric": [
        "ROC_AUC",
        "AP"
    ],

    "median": [
        roc_median,
        ap_median
    ],

    "Q1": [
        roc_q1,
        ap_q1
    ],

    "Q3": [
        roc_q3,
        ap_q3
    ]

})


performance_summary.to_csv(

    os.path.join(

        RESULT_DIR,

        "XGB_performance_summary.csv"

    ),

    index=False

)


# ============================================================
# 37. Pooled OOF ROC
#
# 仅作visualization
#
# 正式模型性能：
# 仍然优先报告9个donor median [IQR]
# ============================================================

y_oof = (

    predictions_df[
        "true_label"
    ]

    .values

)


prob_oof = (

    predictions_df[
        "predicted_probability"
    ]

    .values

)


fpr, tpr, _ = roc_curve(

    y_oof,

    prob_oof

)


pooled_roc = (

    roc_auc_score(

        y_oof,

        prob_oof

    )

)


plt.figure(
    figsize=(
        6,
        6
    )
)


plt.plot(

    fpr,

    tpr,

    linewidth=2,

    label=(
        f"XGBoost AUC = "
        f"{pooled_roc:.3f}"
    )

)


plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--"

)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "XGBoost\nLeave-One-Donor-Out ROC"
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "XGB_LODO_ROC.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 38. Pooled Precision-Recall
# ============================================================

precision, recall, _ = (

    precision_recall_curve(

        y_oof,

        prob_oof

    )

)


pooled_ap = (

    average_precision_score(

        y_oof,

        prob_oof

    )

)


plt.figure(
    figsize=(
        6,
        6
    )
)


plt.plot(

    recall,

    precision,

    linewidth=2,

    label=(
        f"XGBoost AP = "
        f"{pooled_ap:.3f}"
    )

)


plt.xlabel(
    "Recall"
)

plt.ylabel(
    "Precision"
)

plt.title(
    "XGBoost\n"
    "Leave-One-Donor-Out Precision-Recall"
)

plt.legend(
    loc="lower left"
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "XGB_LODO_PR.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 39. Feature stability plot
# ============================================================

plot_df = (

    feature_summary

    .head(
        20
    )

    .copy()

)


plot_df = (

    plot_df

    .sort_values(

        "top20_count",

        ascending=True

    )

)


plt.figure(
    figsize=(
        7,
        8
    )
)


plt.barh(

    plot_df[
        "gene"
    ],

    plot_df[
        "top20_count"
    ]

)


plt.xlabel(
    "Number of outer folds in Top 20"
)

plt.ylabel(
    "Gene"
)

plt.title(
    "XGBoost SHAP Feature Stability"
)

plt.xlim(
    0,
    len(donors)
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "XGB_Top20_feature_stability.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 40. Top20 SHAP magnitude plot
# ============================================================

shap_plot_df = (

    feature_summary

    .head(
        20
    )

    .sort_values(

        "mean_abs_SHAP",

        ascending=True

    )

)


plt.figure(
    figsize=(
        7,
        8
    )
)


plt.barh(

    shap_plot_df[
        "gene"
    ],

    shap_plot_df[
        "mean_abs_SHAP"
    ]

)


plt.xlabel(
    "Mean absolute SHAP value"
)

plt.ylabel(
    "Gene"
)

plt.title(
    "XGBoost Donor-Balanced SHAP Importance"
)

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "XGB_Top20_mean_abs_SHAP.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 41. Donor-level performance
# ============================================================

performance_plot = (

    outer_metrics_df

    .sort_values(
        "ROC_AUC"
    )

    .copy()

)


x = np.arange(
    len(
        performance_plot
    )
)


width = 0.35


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.bar(

    x - width / 2,

    performance_plot[
        "ROC_AUC"
    ],

    width,

    label=
        "ROC-AUC"

)


plt.bar(

    x + width / 2,

    performance_plot[
        "AP"
    ],

    width,

    label=
        "AP"

)


plt.xticks(

    x,

    performance_plot[
        "test_donor"
    ],

    rotation=45,

    ha="right"

)


plt.ylim(
    0,
    1
)

plt.ylabel(
    "Performance"
)

plt.xlabel(
    "Held-out donor"
)

plt.title(
    "XGBoost Performance by Held-out Donor"
)

plt.legend()

plt.tight_layout()


plt.savefig(

    os.path.join(

        FIG_DIR,

        "XGB_outer_donor_performance.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 42. Print stable genes
# ============================================================

stable_genes = (

    feature_summary[

        feature_summary[
            "stable_top20"
        ]

    ]

)


print("\n")

print(
    "=" * 90
)

print(
    "Stable XGBoost genes "
    "(Top20 in >= 6/9 folds)"
)

print(
    "=" * 90
)


if len(
    stable_genes
) == 0:

    print(

        "No genes reached "
        ">= 6/9 Top20 stability."

    )


else:

    print(

        stable_genes[
            [
                "gene",
                "top20_count",
                "mean_abs_SHAP",
                "median_abs_SHAP",
                "mean_HTS_minus_LTS",
                "expression_direction",
                "mean_expression_direction_consistency"
            ]
        ]

        .to_string(
            index=False
        )

    )


# ============================================================
# 43. Done
# ============================================================

print(
    "\nAnalysis completed."
)


print(

    f"\nFigures saved to: "
    f"{FIG_DIR}/"

)


print(

    f"Results saved to: "
    f"{RESULT_DIR}/"

)
```
